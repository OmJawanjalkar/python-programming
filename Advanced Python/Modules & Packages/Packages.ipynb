## Packages in Python 

  Packages are used to organize multiple modules in a structured way.

  - In real-world projects:

    Large programs are divided into packages

## What is a Package?

- A package is a folder that contains:

  Python modules (.py files)
  - sub-packages
  - __init__.py

<!-- Package = Collection of Modules -->

## Package = Collection of Modules

mypackage/  # this is mypackage = Calculator
│
├── __init__.py
├── add.py
└── sub.py

# init.py
Used to:
 - identify folder as package
 - initialize package

# Advantages of Packages
   - Better code organization
   - Reusable code
   - Easy maintenance
   - Supports large projects

# Difference Between Module and Package

       - Module is  Single .py file
       - Package Folder containing modules

# Important Points

   - Packages contain modules

    - Used in large real-world projects
