```python
try:

      ## Code block where exception can occur
    a = 1
    b = "s"
    c = a + b

except NameError:
  print("The user may have not defined the variable")

except TypeError:
     print("Try to make data type similar")

except:
       print("Some Problem may occure")
```

```python
## Try Else Block

try:

      ## Code block where exception can occur
    a = int(input("Enter the Number 1: "))
    b = int(input("Enter the Number 2: "))
    c = a / b
    d = a * b
    f = a + b

except NameError:
    print("The user may have not defined the variable")

except TypeError:
    print("Try to make data type similar")

except ZeroDivisionError:
    print("Please provide number greater than zero")

except Exception as ex:
    print(ex)

else:
    print(c)
    print(d)
    print(f)
```

finally Block
Runs always.
Used for:

closing files
database cleanup
releasing resources

```python
## Try else and finally

try:

      ## Code block where exception can occur
    a = int(input("Enter the Number 1: "))
    b = int(input("Enter the Number 2: "))
    c = a / b
    d = a * b
    f = a + b

except NameError:
    print("The user may have not defined the variable")

except TypeError:
    print("Try to make data type similar")

except ZeroDivisionError:
    print("Please provide number greater than zero")

except Exception as ex:
    print(ex)

else:
  print(c)

finally:
  print("Execution is Done!!")
```
