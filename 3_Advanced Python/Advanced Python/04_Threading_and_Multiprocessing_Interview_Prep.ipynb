# Top 20 Multithreading & Multiprocessing Interview Questions (Python)

## 1. What is a Process?

**Answer:** A process is an independent program in execution with its own memory space.

---

## 2. What is a Thread?

**Answer:** A thread is a lightweight unit of execution inside a process that shares memory and resources.

---

## 3. What is the Difference Between a Process and a Thread?

| Process         | Thread          |
| --------------- | --------------- |
| Own Memory      | Shared Memory   |
| Heavyweight     | Lightweight     |
| Slower Creation | Faster Creation |
| More Isolation  | Less Isolation  |

---

## 4. What is Multithreading?

**Answer:** Running multiple threads concurrently within a single process.

---

## 5. What is Multiprocessing?

**Answer:** Running multiple processes simultaneously, each with its own memory.

---

## 6. Why Do We Use Multithreading?

**Answer:**

* API Calls
* Web Scraping
* File Reading
* Database Queries

---

## 7. Why Do We Use Multiprocessing?

**Answer:**

* Machine Learning Training
* Heavy Computations
* Image Processing
* Feature Engineering

---

## 8. What is the GIL?

**Answer:** GIL (Global Interpreter Lock) allows only one thread to execute Python bytecode at a time.

---

## 9. Why is GIL Important?

**Answer:** Because it limits true parallel execution of Python threads.

---

## 10. Does Multithreading Improve CPU-Bound Tasks?

**Answer:** No. Due to the GIL, CPU-intensive tasks usually do not benefit from multithreading.

---

## 11. Does Multithreading Improve I/O-Bound Tasks?

**Answer:** Yes. It is very effective for I/O operations like API calls and file handling.

---

## 12. What is start()?

**Answer:** `start()` begins the execution of a thread or process.

```python
t.start()
```

---

## 13. What is join()?

**Answer:** `join()` waits until a thread or process completes execution.

```python
t.join()
```

---

## 14. What is ThreadPoolExecutor?

**Answer:** A high-level API used to manage multiple threads efficiently.

```python
from concurrent.futures import ThreadPoolExecutor
```

---

## 15. What is ProcessPoolExecutor?

**Answer:** A high-level API used to manage multiple processes efficiently.

```python
from concurrent.futures import ProcessPoolExecutor
```

---

## 16. What is a Lock?

**Answer:** A lock prevents multiple threads from accessing shared data simultaneously.

```python
lock = threading.Lock()
```

---

## 17. What is Race Condition?

**Answer:** A race condition occurs when multiple threads modify shared data at the same time, causing unpredictable results.

---

## 18. Why Doesn't Multiprocessing Output Show in Jupyter?

**Answer:** Child processes run separately from the IPython kernel, so their output may not appear in notebook cells.

---

## 19. When Should You Use Multithreading vs Multiprocessing?

### Multithreading

* API Calls
* Web Scraping
* Reading Files
* Database Operations

### Multiprocessing

* Model Training
* Numerical Computation
* Image Processing
* Data Transformation

---

## 20. What is the Main Advantage of Multiprocessing?

**Answer:** It bypasses the GIL because each process has its own Python interpreter and memory space.

# One-Line Interview Definitions

✅ **Process:** Independent program with its own memory.

✅ **Thread:** Lightweight execution unit inside a process.

✅ **Multithreading:** Running multiple threads concurrently.

✅ **Multiprocessing:** Running multiple processes simultaneously.

✅ **GIL:** Allows only one thread to execute Python bytecode at a time.

✅ **Lock:** Protects shared resources from concurrent access.

✅ **Race Condition:** Unexpected behavior caused by simultaneous access to shared data.
