### Topic 1: Functions are Objects

      In Python, functions are treated like objects.

      This means a function can:

          - Be assigned to a variable
          - Be passed as an argument to another function
          - Be returned from another function
          - Be stored in data structures (list, dictionary, etc.)

### Example 1: Assign Function to a Variable

````python
def hello():
    print("Hi")

a = hello

a()
````

### Why is this useful?

        Because we can pass functions around just like variables.

````python
def greet():
    print("Hello")

def execute(func):
    func()

execute(greet)
````

 Here:

    - greet is given to execute.
    - execute runs it using func().

# 2. Nested Functions (Function Inside Function)

    A nested function means:

    A function created inside another function.

````python
def outher():

    def inner():
        print(" I am Inner function")

    inner()

outher()
````

Just like an inner function exists inside an outer function.

# 3. Returning Functions

      A function can return another function as a value.

````python
def outer():

   def inner():
       print(" I am Inner function")
   return inner


fun = outer()

fun()
````

### How It Works

          outer()
            │
            ▼
          Creates inner()
            │
            ▼
          Returns inner function
            │
            ▼
          func = inner
            │
            ▼
          func()
            │
            ▼
          Hello from Inner Function

# 4. Closures

## Definition

A **Closure** is a function that remembers the variables from its outer function even after the outer function has finished executing.

### Example

```python
def outer():

    message = "Hello"

    def inner():
        print(message)

    return inner

func = outer()

func()
```

### Output

```python
Hello
```

---

## How It Works

```text
outer()
   │
   ▼
message = "Hello"
   │
   ▼
Creates inner()
   │
   ▼
Returns inner()
   │
   ▼
outer() finishes
   │
   ▼
inner() still remembers message
   │
   ▼
Hello
```

---

## Real-Life Example

Think of a teacher writing a note for a student.

```text
Teacher (outer function)
       │
       ▼
Writes a Note (message)
       │
       ▼
Gives it to Student (inner function)
       │
       ▼
Teacher Leaves
       │
       ▼
Student Still Has the Note
```

Just like the student remembers the note, the inner function remembers the outer variable.

---

## Key Point

A Closure = **Nested Function + Returning Function + Remembering Outer Variables**

---

## One-Line Interview Answer

**A Closure is a function that remembers and can access variables from its outer scope even after the outer function has finished execution.**


# Decorators

### Scenario: Login System

````python
def login_requird(func):

   def wrapper():
      print("Check Login......")
      func()

   return wrapper


def dashbord():
   print("welcome to dashbord")

Dashbord = login_requird(dashbord)
Dashbord()
````

# How a Decorator Works

## Step 1

```python
def dashboard():
    print("Welcome to Dashboard")
```

This is the original function.

---

## Step 2

```python
dashboard = login_required(dashboard)
```

Here:

* `dashboard` is passed to `login_required()`
* `func = dashboard`
* `login_required()` returns `wrapper`
* Now `dashboard` points to `wrapper`

```text
dashboard
    │
    ▼
login_required(dashboard)
    │
    ▼
returns wrapper
    │
    ▼
dashboard = wrapper
```

---

## Step 3

```python
dashboard()
```

Actually calls:

```python
wrapper()
```

---

## Step 4

Inside `wrapper()`:

```python
print("Check Login......")
```

### Output

```python
Check Login......
```

Then:

```python
func()
```

calls the original `dashboard()` function.

### Output

```python
Welcome to Dashboard
```

---

## Final Output

```python
Check Login......
Welcome to Dashboard
```

---

## One-Line Summary

**The decorator adds "Check Login..." before running the original `dashboard()` function without changing the original function's code. 🚀**

---

# With @ Syntax

````python
def login_required(func):

    def wrapper():
        print("Check Login...")
        func()

    return wrapper


@login_required
def dashboard():
    print("Welcome to Dashboard")

dashboard()
````

# Decorators with Arguments

    A decorator with arguments allows you to pass extra information to the decorator.

````python
def login_required(role):

    def decorator(func):

        def wrapper():
            print(f"Checking {role} Login...")
            func()

        return wrapper

    return decorator


@login_required("Admin")
def dashboard():
    print("Welcome to Dashboard")

dashboard()
````

# Simple Flow

```text
@login_required("Admin")
        ↓
role = "Admin"
        ↓
dashboard is passed to decorator
        ↓
wrapper is returned
        ↓
dashboard() calls wrapper()
```

## Why Use It?

When you want to customize the decorator.

### Example

```python
@login_required("Admin")
@login_required("User")
@permission("Manager")
```

Each function can have different settings.

## Interview Definition

A decorator with arguments is a decorator that accepts additional parameters before decorating a function. It usually requires three nested functions. 🚀

# Top 10 Decorator Interview Questions (Python)

## 1. What is a Decorator?

### Answer:

A decorator is a function that takes another function as input, adds extra functionality, and returns a new function.

---

## 2. Why are Decorators Used?

### Answer:

To add functionality without modifying the original function code.

### Examples:

* Logging
* Authentication
* Authorization
* Timing
* Caching

---

## 3. What Concepts Are Required to Understand Decorators?

### Answer:

* Functions are Objects
* Nested Functions
* Returning Functions
* Closures

---

## 4. What Does @decorator Mean?

### Answer:

```python
@decorator
def func():
    pass
```

is equivalent to:

```python
def func():
    pass

func = decorator(func)
```

---

## 5. What is a Closure?

### Answer:

A closure is an inner function that remembers variables from the outer function even after the outer function has finished execution.

---

## 6. Difference Between Decorator and Closure?

### Answer:

**Closure:**

* Remembers outer variables.

**Decorator:**

* Uses closures to modify function behavior.

---

## 7. What is the Difference Between return wrapper and return wrapper()?

### Answer:

```python
return wrapper
```

Returns the function itself.

```python
return wrapper()
```

Executes the function immediately and returns its result.

---

## 8. How Many Functions Are Needed in a Basic Decorator?

### Answer:

Two:

```python
def decorator(func):

    def wrapper():
        pass

    return wrapper
```

---

## 9. How Many Functions Are Needed in a Decorator with Arguments?

### Answer:

Three:

```python
def outer(arg):

    def decorator(func):

        def wrapper():
            pass

        return wrapper

    return decorator
```

---

## 10. Give a Real-World Use Case of Decorators

### Answer:

Login Authentication:

```python
@login_required
def dashboard():
    pass
```

The decorator checks whether the user is logged in before executing the function.
