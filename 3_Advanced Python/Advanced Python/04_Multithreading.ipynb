# What is a Process?

 A process is an independent program that is running.

    - Examples
    - Chrome Browser
    - VS Code
    - Spotify

 Each of these runs as a separate process with its own memory.

      Example
      - Process 1 -> Chrome
      - Process 2 -> VS Code
      - Process 3 -> Spotify

# What is a Thread?

    A thread is the smallest unit of execution inside a process.

    A process can have one or more threads.

    Example

    Inside Chrome:

        Chrome Process
            ├── Thread 1 -> User Interface
            ├── Thread 2 -> Video Playback
            ├── Thread 3 -> Download Files
            └── Thread 4 -> Network Requests

    All threads share the same memory of the process.

# Interview Definition

### Process

A **Process** is an independent program in execution with its own memory space.

**Example:** Running Chrome, VS Code, or Python are separate processes.

---

### Thread

A **Thread** is a lightweight unit of execution within a process that shares the process's memory and resources.

**Example:** Multiple tabs running inside the same browser process.

---

## One-Line Interview Answer

**Process:** An independent program with its own memory.

**Thread:** A lightweight execution unit inside a process that shares memory with other threads.

# Practical Implemention of the multithreding

````python
import threading
import time

def print_numbers():
    for number in range(5):
        time.sleep(2)
        print(f"Numbers: {number}")
      
def print_letters():
   for letters in "abcd":
      time.sleep(2)
      print(f"Letters: {letters}")
      
thread1 = threading.Thread(target=print_numbers)
thread2 = threading.Thread(target=print_letters)

      


t= time.time()
## Starting the threds 
thread1.start()
thread2.start()

thread1.join()
thread2.join()

finished_time = time.time()-t
print(finished_time)
      

````

# Example 2

````python
import threading
import time


def fetch_data(api_name):
  print(f"fetching from {api_name}")
  time.sleep(2)
  print(f"Data recived from {api_name}")
  

start = time.time()
  
t1 = threading.Thread(target=fetch_data, args=("Weather API",))
t2 = threading.Thread(target=fetch_data, args=("Stocks API",))
t3 = threading.Thread(target=fetch_data, args=("News API",))

  
t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()
  
end = time.time()

print(f"\nTotal Time: {end-start:.2f} seconds")
  
  
````

## Multiprocessing (Short Definition)

    Multiprocessing is a technique where multiple processes run simultaneously using different CPU cores to perform heavy computations faster.

    Why Use It?

    It bypasses Python's GIL, allowing true parallel execution.


# Practilcal Implemention of the Multiprocessing 

````python
import multiprocessing
import time 

def square_number():
  for i in range(5):
    time.sleep(1)
    print(f"Square {i*i}")

def cube_number():
  for i in range(5):
    time.sleep(1.5)
    print(f"Cube {i*i*i}")
    
## Create two process
if __name__ == "__main__":

    p1 = multiprocessing.Process(target=square_number)
    p2 = multiprocessing.Process(target=cube_number)

    ## Start the process 
    p1.start()
    p2.start()

    ## Wait for the process to complet 

    p1.join()
    p2.join()
    
    print("Both processes completed")
    
````

# Why Doesn't Multiprocessing Output Show in Jupyter?

## Short Answer

Jupyter Notebook runs code inside a special **IPython Kernel**.

When Multiprocessing creates a new process:

```text
Parent Process → Jupyter Kernel
       │
       ▼
Child Process → Separate Python Process
```

The child process output is not always connected to the notebook output cell.

---

## Why Does This Happen?

Each child process runs independently with its own memory and execution environment.

So:

```python
print("Hello")
```

inside a child process may not appear in the notebook output.

---

## Key Point

* Threads share the same process, so their output usually appears in Jupyter.
* Processes are separate programs, so their output may not be displayed correctly.

---

## One-Line Interview Answer

**Multiprocessing output may not appear in Jupyter because child processes run separately from the IPython kernel and are not directly connected to the notebook output cell.**

# ThreadPoolExecutor and ProcessPoolExecutor


      These are modern ways to perform multithreading and multiprocessing in Python.

      Import:

      from concurrent.futures import ThreadPoolExecutor
      from concurrent.futures import ProcessPoolExecutor

# Prctical implemention of ThreadPoolExecutor

````python
from concurrent.futures import ThreadPoolExecutor
import time

def Task(n):
  time.sleep(1.5)
  return f"Task {n} Completed"

with ThreadPoolExecutor(max_workers=3) as executor:
  results = executor.map(Task, [1, 2, 3, 4, 5])
  
for result in results:
  print(result)
````

# Important Multithreading Concepts for Data Science

## 1. Process vs Thread

### Process

A process is an independent program in execution with its own memory space.

### Thread

A thread is a lightweight unit of execution within a process that shares the process's memory and resources.

### Data Science Example

```text
Python Process
      │
      ├── Thread 1 → Download Data
      ├── Thread 2 → Read CSV
      └── Thread 3 → Fetch API Data
```

---

## 2. Thread Creation (`threading.Thread`)

Used to run multiple tasks concurrently.

### Example

```python
import threading

t = threading.Thread(target=fetch_data)
t.start()
```

### Use Cases

* API Calls
* Web Scraping
* Data Collection
* File Reading

---

## 3. start() and join()

### start()

Starts a thread.

```python
t.start()
```

### join()

Waits for a thread to finish.

```python
t.join()
```

### Data Science Example

```text
Download Dataset A
Download Dataset B
Download Dataset C
        │
        ▼
Wait until all downloads finish
```

---

## 4. GIL (Global Interpreter Lock)

### What is GIL?

Python threads cannot execute Python bytecode in true parallelism.

### Why is it Important?

For CPU-intensive tasks such as:

* Machine Learning Training
* Heavy Calculations
* Matrix Operations

Multithreading may not improve performance significantly.

### Better Alternatives

* Multiprocessing
* NumPy
* PyTorch
* TensorFlow
* GPUs

---

## 5. ThreadPoolExecutor

Used to manage multiple threads efficiently.

### Example

```python
from concurrent.futures import ThreadPoolExecutor

with ThreadPoolExecutor(max_workers=5) as executor:
    executor.map(fetch_data, urls)
```

### Use Cases

* Downloading Datasets
* API Calls
* Web Scraping
* Reading Multiple Files

---

## 6. Lock

A Lock prevents multiple threads from modifying shared data at the same time.

### Example

```python
import threading

lock = threading.Lock()

with lock:
    shared_variable += 1
```

### Use Cases

* Updating Counters
* Logging Systems
* Shared Caches

---

## 7. Multithreading vs Multiprocessing

### Multithreading

Best For:

* API Calls
* Web Scraping
* Reading Files
* Database Queries

### Multiprocessing

Best For:

* Model Training
* Feature Engineering
* Heavy Computation
* Image Processing

---

## 8. Real Data Science Applications

### Data Collection

Fetch data simultaneously from:

* Weather APIs
* Stock Market APIs
* News APIs
* Social Media APIs

```text
Weather API
Stock API
News API
      │
      ▼
Fetch All Simultaneously
```

# Quick Revision

✅ Process → Independent program with its own memory.

✅ Thread → Lightweight worker inside a process.

✅ `start()` → Starts a thread.

✅ `join()` → Waits for thread completion.

✅ GIL → Prevents true parallel execution of Python threads.

✅ ThreadPoolExecutor → Simplifies multithreading.

✅ Lock → Protects shared resources.

✅ Multithreading → Best for I/O-bound tasks.

✅ Multiprocessing → Best for CPU-bound tasks.
