# 1. Iterable

    An Iterable is any object that can be looped over using a for loop.

### Examples of Iterables

````python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_string = "Python"
my_set = {1, 2, 3}
````

    All of these can be used in a for loop:

````python
for item in my_list:
    print(item)
````

### Simple Definition

    An object that contains data and can produce an iterator is called an Iterable.

# 2. Iterator

    An Iterator is an object that returns one element at a time.

    It keeps track of the current position while traversing the data.

### Creating an Iterator

````python
my_list = [10, 20, 30]

its = iter(my_list)

print(next(its))
print(next(its))
print(next(its))
````

````python
my_list = [1, 2, 3]

iterator = iter(my_list)

print(next(iterator))  # Output: 1
# print(next(iterator))  # Output: 2
# print(next(iterator))  # Output: 3
````

 - What is StopIteration?
      - StopIteration is an exception raised when an iterator has no more elements to return.
      - When does it occur?

It occurs when next() is called after all elements of an iterator have been exhausted.

````python
try:
    print(next(iterator))
except StopIteration:
    print("No more items to iterate.")
````

````python
My_tuple = (1, 2, 3)

iterator_1 = iter(My_tuple)

print(next(iterator_1))
print(next(iterator_1))
print(next(iterator_1))
````

````python
my_list = [1, 2, 3, 4, 5]

it = iter(my_list)

while True:
    try:
        print(next(it))
    except StopIteration:
        break
````

# Generator

        A Generator is a special function that gives values one by one whenever you ask for them.

        Instead of creating all values at once, it creates them only when needed.

````python
def number():
   yield 1
   yield 2
   yield 3

gen = number()

print(next(gen)) # Output: 1
print(next(gen)) # Output: 2
print(next(gen)) # Output: 3
````

### What is happening?

    numbers() creates a generator object.
    next(gen) asks for the next value.
    Generator gives one value at a time.

### Think of it like a vending machine:

    You press a button (next()).
    It gives one item (yield).
    Press again, get the next item.

### yield Keyword

      yield is used inside a generator function.

      It works like return, but there is one important difference:

return

            def test():
            return 1
            After returning, the function ends.

### yield


````python
def test():
    yield 1
    yield 2
    yield 3
````

      The function pauses after each yield and continues from the same place when next() is called again

# How a Generator Works

```text
Generator Function
       │
       ▼
yield 1 ──► next() ──► 1
       │
       ▼
yield 2 ──► next() ──► 2
       │
       ▼
yield 3 ──► next() ──► 3
       │
       ▼
StopIteration
```

### Explanation

1. Generator function starts execution.
2. `yield 1` pauses the function and returns `1`.
3. `next()` resumes execution from where it stopped.
4. `yield 2` returns `2`.
5. Another `next()` resumes execution.
6. `yield 3` returns `3`.
7. After the last value, the generator has nothing left to return.
8. Calling `next()` again raises:

```python
StopIteration
```

### Key Point

**A generator remembers its state and resumes execution from the last `yield` statement.**

# 1. Reading Large CSV Files

````python
def read_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
          yield line
````

````python
data = read_file('om.txt')

for row in data:
    print(row)
````

   ###  Benefit:

       - Reads one row at a time.
       - Very memory efficient.

# Top 10 Generator & Iterator Interview Questions (Python)

## 1. What is an Iterable?

**Answer:** An iterable is an object that can be looped over and can create an iterator using `iter()`.

### Examples

* List
* Tuple
* String
* Set
* Dictionary

---

## 2. What is an Iterator?

**Answer:** An iterator is an object that returns elements one by one using `next()`.

```python
it = iter([1, 2, 3])

print(next(it))
```

---

## 3. What is the Difference Between Iterable and Iterator?

| Iterable                     | Iterator                  |
| ---------------------------- | ------------------------- |
| Stores Data                  | Traverses Data            |
| Uses `iter()`                | Uses `next()`             |
| Cannot use `next()` directly | Can use `next()` directly |

---

## 4. What Does iter() Do?

**Answer:** `iter()` converts an iterable into an iterator.

```python
my_list = [1, 2, 3]

it = iter(my_list)
```

---

## 5. What Does next() Do?

**Answer:** `next()` returns the next element from an iterator.

```python
print(next(it))
```

---

## 6. What is a Generator?

**Answer:** A generator is a special iterator that generates values one at a time using `yield`.

```python
def gen():
    yield 1
    yield 2
```

---

## 7. What is yield?

**Answer:** `yield` returns a value and pauses the function, allowing it to resume later from the same point.

```python
def demo():
    yield 10
```

---

## 8. Difference Between return and yield?

| return            | yield                   |
| ----------------- | ----------------------- |
| Ends Function     | Pauses Function         |
| Returns One Value | Returns Multiple Values |
| No State Saved    | State Saved             |

---

## 9. Why are Generators Memory Efficient?

**Answer:** Generators create values only when needed (Lazy Evaluation) instead of storing all values in memory.

```python
def numbers():
    for i in range(1000000):
        yield i
```

---

## 10. What Exception is Raised When an Iterator is Exhausted?

**Answer:** `StopIteration`

```python
it = iter([1])

print(next(it))
print(next(it))
```

### Output

```python
StopIteration
```

# Bonus Interview Questions

## Is Every Iterable an Iterator?

**Answer:** No.

```python
my_list = [1, 2, 3]

next(my_list)   # Error
```

A list is an Iterable but not an Iterator.

---

## Is Every Generator an Iterator?

**Answer:** Yes.

```python
def gen():
    yield 1

g = gen()

print(next(g))
```

Generators support both `iter()` and `next()`, so every generator is an iterator.
