# 1. What is Inheritance?

 ### Inheritance is an OOP concept that allows one class to acquire the properties and methods of another class.

     - The existing class is called the Parent Class (Base Class/Super Class).
     - The new class is called the Child Class (Derived Class/Sub Class).

````python
class Animal():
    def eat(self):
        print("Animal can eat")

class Dog(Animal):
    def Dog_eat(self):
        print("Dog eat meat")

D = Dog()
D.eat()
D.Dog_eat()
D.eat()
````

# 2. Single Inheritance

When one child class inherits from one parent class, it is called Single Inheritance.

    Structure
    Parent
      ↑
      |
    Child

````python
class Person():
    def show_name(self):
        print("Person name is jack")

class Student(Person):
    def show_course(self):
        print("Student course is python")
        
S = Student()
S.show_name()
S.show_course()
````

### 3. Multilevel Inheritance
  
  When a class inherits from another class, and then a third class inherits from that child class, it is called Multilevel Inheritance.

  In simple words:

    - Grandparent → Parent → Child

    - The child class can access the properties and methods of both the parent and grandparent classes.

````python
class A:
    pass

class B(A):
    pass

class C(B):
    pass
````

### Example of multiple inheritance

````python
class grandfather:
    def show_grandfather(self):
        print("Grandfather name is Bhimrao")

class father(grandfather):
    def show_father(self):
        print("Father name is Ramkrushna")

    def Profession(self):
        print("Father profession is Teacher")

class son(father):
    def show_son(self):
        print("Son name is Rahul")

S =  son()
S.show_grandfather()
S.show_father()
S.show_son()
S.Profession()
````

### super() Function in Python
    - What is super()?

    super() is a built-in Python function used to access methods and constructors of the parent class from the child class.

    It is most commonly used to call the parent class's __init__() constructor.

### Example 1: Calling Parent Constructor

````python
class Person:
    def __init__(self):
        print("Person Constructor")

class Student(Person):
    def __init__(self):
        super().__init__()
        print("Student Constructor")

s = Student()
````

    Student Constructor
          ↓
    super().__init__()
          ↓
    Person Constructor

````python
class Employee:
    def __init__(self, name):
        self.name = name
        print("Employee Constructor")

class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language
        print("Developer Constructor")

d = Developer("Om", "Python")

print(d.name)
print(d.language)
````

### 4. Multiple Inheritance

    When a child class inherits from more than one parent class, it is called Multiple Inheritance.

    Structure
    Parent1      Parent2
        \          /
        \        /
          Child

    The child class can access the methods and attributes of both parent classes.

````python
## Basic syntax of multiple inheritance

class Parent1:
    pass

class Parent2:
    pass

class Child(Parent1, Parent2):
    pass
````

````python
class father:
   def skill(self):
      print("Father skill is Programming")

class mother:
    def talent(self):
       print("Mother talent is Singing")

class Child(father, mother):             # Child class is inheriting from both father and mother class happen the multiple inheritance
    def  child_skill(self):
        print("Child skill is Gaming")


C = Child()
C.skill()
C.talent()
C.child_skill()
````

### Using super() in Multiple Inheritance

````python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        super().show()
        print("B")

class C(B):
    def show(self):
        super().show()
        print("C")

obj = C()
obj.show()
````

### Constructor Example

````python
class Father:
    def __init__(self):
        print("Father Constructor")

class Mother:
    def __init__(self):
        print("Mother Constructor")

class Child(Father, Mother):
    pass

obj = Child()
````

### Only the first parent's constructor is called automatically because of MRO.

### Calling Both Parent Constructors

````python
class father:
    def __init__(self):
        print("Father Constructor")

class mother:
    def __init__(self):
        print("Mother Constructor")

class child(father, mother):
    def __init__(self):
        father.__init__(self)   # calling constructor of father class
        mother.__init__(self)   # calling constructor of mother class


obj = child()
````

# 5. Hierarchical Inheritance

### Definition

When multiple child classes inherit from the same parent class, it is called **Hierarchical Inheritance**.

### Structure

```text
      Parent
      /    \
     /      \
 Child1   Child2
```

### Example

```python
class Animal:
    pass

class Dog(Animal):
    pass

class Cat(Animal):
    pass
```

### Key Point

**One Parent → Multiple Child Classes**

---

# 6. Hybrid Inheritance

### Definition

A combination of two or more inheritance types (Single, Multiple, Multilevel, Hierarchical) is called **Hybrid Inheritance**.

### Structure

```text
      A
     / \
    B   C
     \ /
      D
```

### Example

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass
```

### Key Point

**Combination of Different Inheritance Types**

# 7. super() Function

### Definition

`super()` is used to access parent class methods and constructors from a child class.

### Example

```python
class Parent:
    def __init__(self):
        print("Parent Constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Constructor")

c = Child()
```

### Output

```python
Parent Constructor
Child Constructor
```

### Most Common Use

```python
super().__init__()
```

### Key Point

**super() reuses parent class functionality and follows Python's MRO automatically.**


# 8 Method Overriding 

## Definition

When a child class defines a method with the same name as a method in the parent class, the child class method overrides the parent class method.

### Example

```python
class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

### Output

```python
Dog Barks
```

### Why?

`Dog` has its own `sound()` method, so Python uses it instead of the parent's method.

---

## Using super()

```python
class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog Barks")

d = Dog()
d.sound()
```

### Output

```python
Animal Sound
Dog Barks
```

---

## Interview Definition

**Method Overriding is when a child class redefines a parent class method with the same name to provide its own implementation.**

---

## Key Points

* Parent and Child have the same method name.
* Child method gets priority.
* Used to modify parent class behavior.
* `super()` can be used to call the parent method.
