# Instance, Class and Static Methods in Python

Introduction

  In OOP, methods are functions defined inside a class.

  Python provides three types of methods:

    -Instance Method
    -Class Method
    -Static Method

  Each method has a different purpose and is used in different situations.

# 1. Instance Method

An instance method is a method that works with object-specific data.
It can access:

   - Instance variables
   - Class variables

Instance methods use the self parameter.

````python
class Student():
    def display(self):
        print("I am a student.")
````

- Example

````python
class Student():
    def __init__(self, name):
        self.name = name


    def display(self):
      print("Name:", self.name)

s1 = Student("Alice")
print(s1.name)
````

Important Points
Uses self
Can access instance variables
Can access class variables
Called using object

Example:

student1.display()

# 2. Class Method

A class method works with class-level data.

It can access:

    - Class variables

Class methods use the cls parameter.

````python
@classmethod
def class_mthod(cls):
    pass
````

Example 1: Accessing Class Variable

````python
class Student:

    college = "ABC College"

    @classmethod
    def show_college(cls):

        print(cls.college)

Student.show_college()
````

Example 2: Modifying Class Variable

````python
class Student():
    college = "ABC University"

    @classmethod
    def show_college(cls, new_college):
        cls.college = new_college
        print("College:", cls.college)

Student.show_college("XYZ University")
````

Important Points
  -  Uses cls
  -  Uses @classmethod
  -  Accesses class variables
  -  Can be called using class name

Example:

Student.show_college()

Example 3: Using Object to Call Class Method

````python
class Student():
   school = "ABC school"

   @classmethod
   def show_school(cls):
       print("School: ", cls.school)

s1 = Student()
s1.show_school()
````

# Real-Life Example of Class Method

  - Suppose every employee gets a salary increment by the same percentage.

````python
class Employee:
    increment = 1.10


    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def change_increment(cls, new_increment):
        cls.increment = new_increment

e1 = Employee("John", 50000)
e1.change_increment(1.20)
print(Employee.increment)
````

# 3. Static Method

A static method is independent of both:

   - Instance variables
   - Class variables

It behaves like a normal function but is placed inside the class for organization.

Uses:

@staticmethod

````python
# Syntax for static method

class ClassName:

    @staticmethod
    def method_name():
        pass
````

# - No self and no cls.

````python
class Calculator:

    @staticmethod
    def square(num):
        return num * num

print(Calculator.square(5))
````

````python
class Student:

    @staticmethod
    def greet():

        print("Welcome")

Student.greet()
````

    Important Points
    Uses @staticmethod
    No self
    No cls
    Cannot access instance variables directly
    Cannot access class variables directly

````python
## Example of static method

class Employee:

    @staticmethod
    def company_policy():
        print("You have to work 8 hours a day.")

Employee.company_policy()
````

# Complete Example


````python
class Student:

    college = "ABC College"

    def __init__(self, name):

        self.name = name

    # Instance Method
    def display(self):

        print("Name:", self.name)

    # Class Method
    @classmethod
    def show_college(cls):

        print("College:", cls.college)

    # Static Method
    @staticmethod
    def greet():

        print("Welcome")

student1 = Student("Om")

student1.display()

Student.show_college()

Student.greet()
````

# Class Methods and Static Methods - Most Frequently Asked Exam & Interview Questions

## 1. What is a Class Method in Python?

**Answer:**
A class method is a method that belongs to the class rather than an object. It is defined using the `@classmethod` decorator and takes `cls` as its first parameter.

---

## 2. What is a Static Method in Python?

**Answer:**
A static method is a method that does not access instance variables or class variables. It is defined using the `@staticmethod` decorator.

---

## 3. What is the difference between `self` and `cls`?

**Answer:**

| self                        | cls                      |
| --------------------------- | ------------------------ |
| Refers to the object        | Refers to the class      |
| Used in instance methods    | Used in class methods    |
| Accesses instance variables | Accesses class variables |

---

## 4. What is the difference between a Class Method and a Static Method?

**Answer:**

| Class Method               | Static Method                          |
| -------------------------- | -------------------------------------- |
| Uses `@classmethod`        | Uses `@staticmethod`                   |
| First parameter is `cls`   | No special parameter                   |
| Can access class variables | Cannot directly access class variables |
| Can modify class variables | Cannot modify class variables directly |

---

## 5. Why do we use `@classmethod`?

**Answer:**
We use `@classmethod` when we need to access or modify class-level data shared by all objects.

---

## 6. Why do we use `@staticmethod`?

**Answer:**
We use `@staticmethod` for utility or helper functions that do not depend on object or class data.

---

## 7. Can a Class Method access class variables?

**Answer:**
Yes. A class method can access and modify class variables using `cls`.

---

## 8. Can a Class Method access instance variables?

**Answer:**
No. A class method cannot directly access instance variables because it works with the class, not an object.

---

## 9. Can a Static Method access class variables?

**Answer:**
No. Static methods do not have access to `cls`.

---

## 10. Can a Static Method access instance variables?

**Answer:**
No. Static methods do not have access to `self`.

---

## 11. Can a Class Method modify class variables?

**Answer:**
Yes. Class methods can modify class variables using `cls`.

---

## 12. Can we call a Class Method without creating an object?

**Answer:**
Yes. Class methods can be called directly using the class name.

```python
Student.show_school()
```

---

## 13. Can we call a Static Method without creating an object?

**Answer:**
Yes. Static methods can be called directly using the class name.

```python
Math.add(10, 20)
```

---

## 14. What are Alternative Constructors?

**Answer:**
Alternative constructors are class methods used to create objects in different ways.

---

## 15. Why are Static Methods called Utility Methods?

**Answer:**
Because they perform helper tasks and do not depend on object or class data.

---

## 16. Explain Instance Method, Class Method, and Static Method.

**Answer:**

| Method Type     | First Parameter | Access Object Data | Access Class Data |
| --------------- | --------------- | ------------------ | ----------------- |
| Instance Method | self            | Yes                | Yes               |
| Class Method    | cls             | No                 | Yes               |
| Static Method   | None            | No                 | No                |
