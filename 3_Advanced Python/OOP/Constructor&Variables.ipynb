## 1 self Keyword
  - What is self?

  - self refers to the current object.

  - When an object calls a method, Python automatically passes that object as the first argument.

```python
class student:

      def greet(self):
         print("Hello, I am a Student")

s1 = student()
s1.greet()
```

Internally Python does:

Student.greet(student1)

Here:

self = student1

 # Why self is Needed?

To access:

- instance variables
- instance methods

inside the class.

```python
class Student:

    def show(self):
        print("Welcome")

student1 = Student()

student1.show()
```

Without self, Python will give an error.
