## What is OOP?

    Object-Oriented Programming (OOP) is a programming paradigm that organizes code into objects and classes. It helps make programs more modular, reusable, and easier to maintain.

    Think of a class as a blueprint and an object as a real thing created from that blueprint.

Example

A blueprint of a car contains:

   - Brand
   - Model
   - Color

The actual car is an object.

## Class and Object


Class

A class is a blueprint or template used to create objects. It defines the attributes (data) and methods (functions) that objects of that class will have.

Definition : A class is a user-defined data type that acts as a blueprint for creating objects.

```python
# Class in Python

class Student:  ## creating a class named Student
  pass
```

```python
class car:
  color = "red"
  brand = "BMW"
print(car.color)
print(car.brand)
```

> Object

An object is an instance of a class. It is a real entity created from the class blueprint and can access the class's attributes and methods.

Definition : An object is an instance of a class that contains actual data and can perform actions defined by the class.

```python
class Student:  ## creating a class named Student
    pass

s1 = Student()  ## creating an object of the class Student
```

```python
class Car:
  color = "red"
  brand = "BMW"

car1 = Car()
print(car1.color)
print(car1.brand)
```

```python
class Students:
  college = "ABC College"

s1 = Students()
s2 = Students()
```

    - Student → Class
    - s1, s2 → Objects
    - college → Class Variable

Key Point:
Class = Blueprint → Object = Real Thing Created from the Blueprint

# What are Attributes?

  Attributes are variables inside a class.

  They store the data or properties of an object.

Class Attributes

```python
class Student:

    name = "Om"   ## class attributes
    age = 21      ## class attributes
```

Instance Attributes
-- Unique for each object.

```python
class Student:

    def __init__(self, name):

        self.name = name    ## self.name is an instance attribute
```

```python
class Student:

    def __init__(self, name):

        self.name = name

student1 = Student("Om")
student2 = Student("Rahul")

print(student1.name)
print(student2.name)
```

| Class Attribute           | Instance Attribute       |
| ------------------------- | ------------------------ |
| Shared by all objects     | Separate for each object |
| Defined directly in class | Defined using self       |
| One copy                  | Multiple copies          |


# What are Methods?

 - Methods are functions inside a class.

 - They define the behavior of an object.

```python
class Student:
    def greet(self):                            ## greet() → Method
        print("Hello, I am a Student")

Student1 = Student()
Student1.greet()
```

# Multiple Methods Example

```python
class Student:

    def study(self):

        print("Studying")

    def exam(self):

        print("Writing Exam")

student1 = Student()

student1.study()
student1.exam()
```

# Complete Example

```python
class Student:

    college = "ABC College"       ## class attribute

    def greet(self):             ## method

        print("Hello Student")

student1 = Student()

print(student1.college)

student1.greet()
```
