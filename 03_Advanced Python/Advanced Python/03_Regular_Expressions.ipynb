# 1. What is Regular Expression (Regex)?

    A Regular Expression (Regex) is a sequence of characters used to search, match, extract, or replace patterns in text.


### 2. re Module in Python

  The re (Regular Expression) module provides functions to work with Regular Expressions (Regex) in Python.


        Search text
        Match patterns
        Extract data
        Replace text
        Split strings

````python
import re
````

| Function        | Purpose                                 |
| --------------- | --------------------------------------- |
| `re.search()`   | Find first match anywhere in the string |
| `re.match()`    | Check match only at the beginning       |
| `re.findall()`  | Return all matches as a list            |
| `re.finditer()` | Return all matches as iterator          |
| `re.sub()`      | Replace matched pattern                 |
| `re.split()`    | Split string using a pattern            |


        1. re.search()

````python
import re

text = "My age is 22"

result = re.search(r"\d+", text)

print(result.group())
````

      2. re.match()

      re.match() checks only at the beginning of the string.

````python
import re

text = "I am a Data Scientist"


result = re.match("I am", text)

print(result.group())
````

      3. re.findall()

      Returns all matches as a list.

````python
import re

text = "I have 2 Dog and 4 Cat "

result = re.findall(r"\d+", text)

print(result)
````

    4. re.finditer()

    Returns match objects one by one.

````python
import re

text = "I have 2 Dog and 4 Cat "

result = re.finditer(r"\d+", text)

for match in result:
   print(match.group())
````

    5. re.sub()

    Replaces matched patterns.

````python
import re

text = "Om is good Boy"

result = re.sub("good","bad", text)

print(result)
````

      6. re.split()

      Splits text using a pattern.

````python
import re

text = "apple, banana, orange"

result = re.split(",", text)

print(result)
````

# 3. Regex Metacharacters

Metacharacters are special characters that have a special meaning in Regex.

They help define patterns instead of matching literal text.

---

## 1. . (Dot)

Matches any single character except a newline.

```python
import re

text = "cat bat hat"

print(re.findall(".at", text))
```

### Output

```python
['cat', 'bat', 'hat']
```

---

## 2. ^ (Caret)

Matches the start of a string.

```python
import re

text = "Python is easy"

print(re.search("^Python", text))
```

### Output

```python
<Match object>
```

---

## 3. $ (Dollar)

Matches the end of a string.

```python
import re

text = "I love Python"

print(re.search("Python$", text))
```

### Output

```python

```

---

## 4. * (Zero or More)

Matches the preceding character 0 or more times.

```python

```

### Output

```python
['a', 'ab', 'abb', 'abbb']
```

---

## 5. + (One or More)

Matches the preceding character 1 or more times.

```python

```

### Output

```python
['ab', 'abb', 'abbb']
```

---

## 6. ? (Zero or One)

Makes the preceding character optional.

```python

```

### Output

```python

```

# 4. Special Sequences in Regex

      Special sequences are predefined shortcuts that represent commonly used character classes.

| Sequence | Meaning             |
| -------- | ------------------- |
| `\d`     | Digit (0-9)         |
| `\D`     | Non-digit           |
| `\w`     | Word character      |
| `\W`     | Non-word character  |
| `\s`     | Whitespace          |
| `\S`     | Non-whitespace      |
| `\b`     | Word boundary       |
| `\B`     | Not a word boundary |


````python
import re

text = "Om_123 @ Python 2026"

print("Digits:", re.findall(r"\d", text))
print("Non-Digits:", re.findall(r"\D", text))
print("Word Characters:", re.findall(r"\w", text))
print("Non-Word Characters:", re.findall(r"\W", text))
print("Whitespaces:", re.findall(r"\s", text))
print("Non-Whitespaces:", re.findall(r"\S", text))
print("Word Boundary:", re.findall(r"\bPython\b", text))
````



# 7. Email Validation

Email validation is a very common Regex interview question.

### Example

```python
import re

email = "om@gmail.com"

pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```

### Output

```python
Valid Email
```

### Explanation

* `^` → Start of string
* `[\w\.-]+` → Username
* `@` → Must contain @
* `[\w\.-]+` → Domain name
* `\.` → Dot
* `\w+` → Extension
* `$` → End of string

---

# 8. Phone Number Validation

### Example

```python
import re

phone = "9876543210"

pattern = r"^[6-9]\d{9}$"

if re.match(pattern, phone):
    print("Valid Phone Number")
else:
    print("Invalid Phone Number")
```

### Output

```python
Valid Phone Number
```

### Explanation

* `[6-9]` → Starts with 6, 7, 8, or 9
* `\d{9}` → Remaining 9 digits
* Total = 10 digits

# 9. Password Validation

### Requirements

* At least 8 characters
* One uppercase letter
* One lowercase letter
* One digit
* One special character

### Example

```python
import re

password = "Python@123"

pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&]).{8,}$"

if re.match(pattern, password):
    print("Strong Password")
else:
    print("Weak Password")
```

### Output

```python
Strong Password
```

### Explanation

* `(?=.*[a-z])` → At least one lowercase letter
* `(?=.*[A-Z])` → At least one uppercase letter
* `(?=.*\d)` → At least one digit
* `(?=.*[@$!%*?&])` → At least one special character
* `.{8,}` → Minimum 8 characters

## Complete NLP Text Cleaning Program

````python
import re

text = """
I love Python!!! 😊
Visit https://python.org
Call me at 9876543210
#AI @OpenAI
"""

# Remove URLs
text = re.sub(r"http\S+", "", text)

# Remove Numbers
text = re.sub(r"\d+", "", text)

# Remove Special Characters
text = re.sub(r"[^a-zA-Z\s]", "", text)

# Remove Extra Spaces
text = re.sub(r"\s+", " ", text).strip()

print(text)
````
