```python
class error(Exception):
  pass

class dobException(error):
  pass
```

```python
year = int(input("Enter the Birth Year : "))
age = 2026 - year

try:
    if  age<=30  & age>=20:
        print("The age is valid. You can apply for the Exam")
    else:
        raise dobException
except dobException:
    print("The age is not valid. You Canoot apply for the Exam ")

else:
  print(age)

```
