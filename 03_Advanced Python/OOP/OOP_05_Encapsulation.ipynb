 # - Encapsulation

# What is Encapsulation?

  Encapsulation is the process of wrapping data and methods into a single unit (class) and restricting direct access to data by using private variables and public methods.

   -  Encapsulation = Data Hiding + Data Protection

# 1. Public Members

Public members can be accessed from anywhere.

````python
class Student:

    def __init__(self):

        self.name = "Om"

student1 = Student()

print(student1.name)
````

 - Public Variable

````python
class Student:

    name = "Om"

````

  - Public Method

````python
class Student:
  def display(self):
    print("Hello World")

# Accessible everywhere.
````

 # 2. Protected Members

    Protected members are intended to be used only inside the class and its child classes.

    Syntax
    
        - _variable_name

    Single underscore (_) is used.

````python
# Examle  

class Student:
   def __init__(self):
        self._name = "Om"

student1 = Student()
print(student1._name)
````

````python
class Employee:
    def __init__(self):
        self._salary = 50000
        self._bonus = 5000
e1 = Employee()
print(e1._salary)
print(e1._bonus)
````

Important Point

        Even though we can access _name outside the class, we should not.

        The underscore is a warning to programmers:

        "This variable is meant for internal use."

# Protected Members with Inheritance

````python
class person:
    def __init__(self):
        self._name = "Om"

class student(person):
    def display(self):
        print("Name:", self._name)

e1 = student()
e1.display()
````

   - Explanation
      -  _name belongs to Person.
      -  Student inherits Person.
      -  Child class can access protected members.

# 3. Private Members

Private members are used when data should not be accessed directly outside the class.

Syntax

   - __variable_name

Double underscore (__) is used.

````python
class Student:
   def __init__(self):
        self.__name = "Om"

student1 = Student()
print(student1.__name)
````

  - Why Error Occurs?

      Python hides private variables using a mechanism called Name Mangling.

       Internally:

           - self.__name

      becomes

           - self._Student__name

      So Python cannot find __name directly.

  - Accessing Private Variables Using Methods

````python
class Student:
   def __init__(self):
        self.__name = "Om"

   def display(self):
       print("Name:", self.__name)

student1 = Student()
student1.display()
````

# 4. Getter Methods

    A Getter Method is used to access (read) private data safely from outside the class.

    Since private variables (__variable) cannot be accessed directly, we use a getter method.

````python
# Syntax

def get_variable(self):
    return self.__variable
````

````python
# Example  of the getters methods

class student:

   def __init__(self):
        self.__name = "Om"

   def get_name(self):
        return self.__name

student1 = student()
print(student1.get_name())
````

      Explanation
      def get_name(self):
          return self.__name
      get_name() is a getter method.
      It returns the value of the private variable.
      Users can read the data without directly accessing it.

````python
## Example of the setters method


class student:
   def __init__(self):
        self.__mark = 85

   def get_mark(self):
        return self.__mark

S1 = student()
print(S1.get_mark())
````

# How It Works

### Step 1

```python
s = Student()
```

`__marks = 85`

### Step 2

Getter method is called:

```python
s.get_marks()
```

### Step 3

Method returns:

```python
return self.__marks
```

## Key Point

* Private variables cannot be accessed directly from outside the class.
* Getter methods provide controlled access to private data.
* They help maintain data security and encapsulation.


# 5. Setter Method
  Definition

  A Setter Method is used to modify or update private data with validation/control.

  Instead of changing the private variable directly, we use a setter method.

````python
# Syntax

def set_variable(self, value):
    self.__variable = value
````

````python
## Example of the setters method

class student:
    def __init__(self):
        self.__mark = 85

    def set_mark(self, mark):
        self.__mark = mark

    def get_mark(self):
        return self.__mark

S1 = student()
S1.set_mark(90)
print(S1.get_mark())

````

# How It Works

### Initial Value

```python
__marks = 85
```

### Setter Called

```python
s.set_marks(95)
```

### Value Updated

```python
self.__marks = 95
```

### Getter Called

```python
s.get_marks()
```

### Output

```python
95
```


## Real Advantage of Setter: Validation

Suppose marks cannot be less than 0 or greater than 100.

````python
class student:
    def __init__(self):
      self.__mark = 0
      
    def set_mark(self, mark):
       if 0 <= mark <= 100:
          self. __mark = mark
       else:
         print("Invalid mark. Please enter a value between 0 and 100.")

    def get_mark(self):
        return self.__mark

s1 = student()
s1.set_mark(94)
print(s1.get_mark())
s1.set_mark(150)
print(s1.get_mark())
````

    Notice that 150 was not accepted because of validation.

### Complete Example

````python
class BankAccount:
    def __init__(self):
        self.__balance = 1000

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative")

account = BankAccount()

print(account.get_balance())

account.set_balance(5000)
print(account.get_balance())

account.set_balance(-100)
````

### Answer:

By using double underscores (`__`) before the variable name.

```python

        self.__marks = 95
```

`__marks` becomes a private variable.

## 9. What is the Difference Between Public, Protected, and Private Variables?

| --------------- | ------ | ------------------------ |
| Public          | name   | Yes                      |
| Protected       | _name  | Yes (by convention)      |
| Private         | __name | No                       |

# Most Frequently Asked Exam Question

## Explain Encapsulation with a Real-Life Example.

### Answer:

A bank account is a real-life example of encapsulation. The account balance is private and cannot be modified directly. Users interact with the balance through methods such as deposit and withdraw.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 1000

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount()

account.deposit(500)

print(account.get_balance())
```

### Output

```python
1500
```
