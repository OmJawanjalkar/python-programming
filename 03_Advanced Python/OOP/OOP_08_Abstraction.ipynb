# 1. What is Abstraction?

    Abstraction is the process of hiding unnecessary implementation details and showing only the essential features to the user.

### In simple words:

          - The user knows what an object does, but not how it does it.

### Example: Car

You use:

    Steering
    Brake
    Accelerator

You don't need to know:

    How the engine works
    How fuel combustion happens

The car hides the implementation details.

Abstraction is used to hide complex implementation details and show only the necessary functionality to the user.

# 3. Abstract Class

An Abstract Class is a class that cannot be instantiated (you cannot create its object directly).

It is used as a blueprint for other classes.

An abstract class may contain:

   - Abstract methods
   - Normal methods
   - ariables

The main purpose of an abstract class is to define a common structure that all child classes must follow.

````python
# Synatx

from abc import ABC, abstractmethod

class Animal(ABC):
    pass
````

## Here:

  - ABC stands for Abstract Base Class.
  - Animal becomes an Abstract Class.

````python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def speak(self):
      pass
````

    Animal is now an abstract class because it contains an abstract method.

Key Points

✅ Cannot create object directly

✅ Acts as blueprint/template

✅ Contains abstract methods

✅ Used to enforce rules on child classes

Interview Answer
    An Abstract Class is a class that cannot be instantiated and is used as a blueprint for other classes. It may contain abstract methods that must be implemented by child classes.

# 4. Abstract Method

    An Abstract Method is a method that is declared in the abstract class but has no implementation.

    It only specifies what should be done, not how it should be done.

    Child classes provide the actual implementation.

````python
@abstractmethod
def speak(self):
  pass
````

 ### Here:

  - @abstractmethod is a decorator.
  - pass means no implementation is provided.

````python
# Complete Example

from abc import ABC, abstractmethod

class Animal(ABC):

   @abstractmethod
   def speak(self):
      pass

class Dog(Animal):
    def speak(self):
      print("Woof!")

class Cat(Animal):
    def speak(self):
      print("Meow!")

dog = Dog()
cat = Cat()

dog.speak()
cat.speak()
````

### Complet Example 

````python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
   def __init__(self, radius):
       self.radius = radius

   def area(self):
       return 3.14 * self.radius ** 2

class Rectangle(Shape):
   def __init__(self, width, height):
        self.width = width
        self.height = height

   def area(self):
        return self.width * self.height

circle = Circle(5)
rectangle = Rectangle(4, 6)
print("Area of Circle:", circle.area())
print("Area of Rectangle:", rectangle.area())
````

### Simple Analogy

    Think of Shape as a company rule:

    "Every shape must know how to calculate its area."

       - Circle follows the rule in its own way.
       -  Rectangle follows the rule in its own way.

    The abstract class defines the rule, and the child classes provide the implementation.

# ABC Module

## What is ABC Module?

**ABC** stands for **Abstract Base Class**.

It is used to create:

* Abstract Classes
* Abstract Methods

```python
from abc import ABC, abstractmethod
```

---

## Why Do We Need It?

We use the ABC module when we want to force child classes to implement specific methods.

---

## Components

### ABC

Used to create an Abstract Class.

```python
class Animal(ABC):
    pass
```

### @abstractmethod

Used to create an Abstract Method.

```python
@abstractmethod
def sound(self):
    pass
```

---

## Key Points

* `ABC` → Creates an Abstract Class.
* `@abstractmethod` → Creates an Abstract Method.
* Forces child classes to implement required methods.
* Helps achieve Abstraction.

---

## Interview Definition

**ABC (Abstract Base Class) is a Python module used to create Abstract Classes and Abstract Methods.**


# 6. Child Class Implementation

## What is Child Class Implementation?

A child class must implement all abstract methods of the abstract class.

### Example

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

### Output

```python
Dog Barks
```

---

## Key Point

* Child class provides the actual implementation of abstract methods.
* Without implementation, the child class object cannot be created.

---

## Interview Definition

**Child Class Implementation is the process of defining all abstract methods inherited from an abstract class.**
