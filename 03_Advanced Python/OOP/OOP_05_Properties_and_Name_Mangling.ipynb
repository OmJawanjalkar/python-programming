# Property Decorators and Name Mangling in Python

## 1. Property Decorators

Before Python introduced Property Decorators, we used Getter and Setter methods like this:

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks

s = Student(85)

print(s.get_marks())

s.set_marks(95)
```

### Problem

The syntax is lengthy:

```python
s.get_marks()
s.set_marks(95)
```

Python provides a cleaner solution using Property Decorators.

---

## What is a Property Decorator?

A property decorator allows a method to be accessed like an attribute.

Instead of:

```python
s.get_marks()
```

We can write:

```python
s.marks
```

This makes code cleaner and more Pythonic.

---

# 2. @property Decorator

## Definition

`@property` converts a method into a getter.

### Example

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    @property
    def marks(self):
        return self.__marks

s = Student(85)

print(s.marks)
```

### Output

```python
85
```

---

## How it Works

### Step 1

Object creation:

```python
s = Student(85)
```

### Step 2

When Python sees:

```python
s.marks
```

it automatically calls:

```python
marks()
```

behind the scenes.

So:

```python
print(s.marks)
```

actually behaves like:

```python
print(s.marks())
```

but without parentheses.

---

## Why Use @property?

### Without Property

```python
student.get_marks()
```

### With Property

```python
student.marks
```

Cleaner and easier to read.

---

# 3. @setter Decorator

## Definition

`@setter` allows controlled modification of a property.

### Example

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Invalid Marks")

s = Student(85)

print(s.marks)

s.marks = 95

print(s.marks)
```

### Output

```python
85
95
```

---

## What Happens Internally?

When Python sees:

```python
s.marks = 95
```

it automatically calls:

```python
marks.setter()
```

which executes:

```python
self.__marks = 95
```

---

## Complete Property Example

```python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, amount):
        if amount > 0:
            self.__salary = amount
        else:
            print("Salary must be positive")

e = Employee(50000)

print(e.salary)

e.salary = 60000

print(e.salary)
```

### Output

```python
50000
60000
```

---

## Property Decorator Interview Question

### Why use @property instead of Getter Methods?

### Answer:

* Cleaner syntax
* More Pythonic
* Easier maintenance
* Supports validation
* Looks like normal attribute access

# 4. Name Mangling

This is one of the most important interview topics.

## What is Name Mangling?

When a variable starts with double underscores (`__`), Python changes its internal name to avoid accidental access.

### Example

```python
class Student:
    def __init__(self):
        self.__marks = 90
```

Python internally changes:

```python
__marks
```

to:

```python
_Student__marks
```

This process is called **Name Mangling**.

---

## Why Does Python Use Name Mangling?

To prevent accidental modification of private variables.

### Example

```python
class Student:
    def __init__(self):
        self.__marks = 90

s = Student()

print(s.__marks)
```

### Output

```python
AttributeError
```

Python cannot find `__marks`.

---

## Let's See the Internal Name

```python
class Student:
    def __init__(self):
        self.__marks = 90

s = Student()

print(s.__dict__)
```

### Output

```python
{'_Student__marks': 90}
```

Notice:

```python
__marks
```

became:

```python
_Student__marks
```

---

## Can We Access It?

Technically Yes.

```python
print(s._Student__marks)
```

### Output

```python
90
```

But this should not be done in real projects.

---

## Important Point

Python does not have true private variables like Java or C++.

Instead, it uses Name Mangling.

So:

```python
__marks
```

is not completely hidden.

It is only renamed.

---

## Name Mangling Example with Inheritance

```python
class Parent:
    def __init__(self):
        self.__data = "Parent Data"

class Child(Parent):
    pass

c = Child()

print(c.__dict__)
```

### Output

```python
{'_Parent__data': 'Parent Data'}
```

Python preserves the parent class's private variable name.

---

## Interview Question

### What is Name Mangling in Python?

### Answer:

Name Mangling is a mechanism where Python automatically changes a private variable name from:

```python
__variable
```

to:

```python
_ClassName__variable
```

to prevent accidental access and naming conflicts.

# Interview One-Liner

**"Modern encapsulation in Python is achieved using @property and @setter, while private variables are protected through Name Mangling (__var → _ClassName__var)."**


### Example 

````python
class student:
    def __init__(self, marks):
      self.__marks = 35

    @property
    def marks(self):
       return self.__marks
    @marks.setter
    def marks(self, value):
        if value < 0:
            self.__marks = 0
        elif value > 100:
            self.__marks = 100
        else:
            self.__marks = value

s1 = student(85)

print(s1.marks)
s1.marks = 105
print(s1.marks)
````
