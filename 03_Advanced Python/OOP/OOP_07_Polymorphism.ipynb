## 1. What is Polymorphism?

  Polymorphism means "many forms".

  In OOP, polymorphism allows the same method name or operation to behave differently for different objects.

  Simple Meaning

   - One action → Multiple behaviors

````python
class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")

class Cow:
    def speak(self):
        print("Moo")

dog = Dog()
cat = Cat()
cow = Cow()

dog.speak()
cat.speak()
cow.speak()
````

````python
print(len("Python"))
print(len([1, 2, 3, 4]))

# The same len() function works with different object types.
````

# 2. Method Overriding

    Method Overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.

    The child class overrides the parent's method.

### Basic Example

````python
class Animal:
    def sound(self):
       print("Animal makes a sound")

class  Dog(Animal):
    def sound(self):
        print("Dog barks")

obj = Dog()
obj.sound()
````

Even though sound() exists in Animal, the Dog version is executed.

### Without Overriding

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass


obj = Dog()
obj.sound()
````

  Since Dog doesn't define sound(), it uses the parent's method.

### Using super()
    Sometimes you want to use the parent's method and then add extra functionality.

````python
class Animal:
    def speak(self):
       print("Animal speaks")

class Dog(Animal):
    def speak(self):
      super().speak()
      print("Dog barks")

obj = Dog()
obj.speak()
````

# Rules of Method Overriding
  ### ✅ Same Method Name

````python
class Parent:
    def show(self):
        pass

class Child(Parent):
    def show(self):
        pass
````

### ✅ Same Parameters (Recommended)

````python
class Parent:
    def show(self, name):
        pass

class Child(Parent):
    def show(self, name):
        pass
````

## 3. Runtime Polymorphism

    Runtime Polymorphism means the method that gets executed is decided while the program is running, not before.

    In Python, Runtime Polymorphism is achieved using Method Overriding.

    Simple Meaning

    When a parent class reference points to a child object, Python decides at runtime which method to call.

      - Same method name → Different behavior → Decided at runtime

### Example

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
     def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
````

````python
class Shape:
    def draw(self):
        print("Drawing Shape")

class Circle(Shape):
    def draw(self):
        print("Drawing Circle")

class Rectangle(Shape):
    def draw(self):
        print("Drawing Rectangle")

shapes = [Circle(), Rectangle()]

for shape in shapes:
    shape.draw()
````

    The same method draw() behaves differently for different objects.

# How Runtime Polymorphism Works

### Step 1

Parent class defines a method.

```python
class Animal:
    def sound(self):
        print("Animal Sound")
```

### Step 2

Child class overrides the method.

```python
class Dog(Animal):
    def sound(self):
        print("Dog Barks")
```

### Step 3

Object is created.

```python
d = Dog()
```

### Step 4

Method is called.

```python
d.sound()
```

### Step 5

Python checks the object's actual type at runtime.

Since `d` is a `Dog` object, Python calls the `Dog` class method.

### Output

```python
Dog Barks
```

# 4. Method Overloading

   - Method Overloading means having multiple methods with the same name but different parameters.

  Note:

     - Python does not support true method overloading like Java or C++.

### Example (Not Possible in Python)

````python
class Demo:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

d = Demo()
print(d.add(1, 2, 3))
````

    Only the last add() method exists. The first one gets overwritten.

### How Python Achieves Overloading

     -  Using default arguments:

````python
class Demo:
    def add(self, a, b, c=0):
        return a + b + c

d = Demo()
print(d.add(1, 3))
print(d.add(1, 2, 3))
````

# Operator Overloading

  Operator Overloading means giving a custom meaning to an operator (+, -, *, etc.) when it is used with objects of a class.

    - Example

````python
print(30 + 20)                 ## Adding two numbers
print("30" + "20")             ## Concatenating strings
````

    The same + operator performs different tasks.

    This is called Operator Overloading.

### Custom Example

````python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

num1 =  Number(10)
num2 = Number(20)

print(num1 + num2)
````

   When n1 + n2 is executed, Python internally calls:

     - n1.__add__(n2)

# Operator Overloading (Short & Simple)

## Interview Definition

**Operator Overloading is the process of giving special meaning to operators for user-defined objects using magic methods like `__add__()`, `__sub__()`, etc.**

---

## Important Magic Methods

| Operator | Magic Method    |
| -------- | --------------- |
| +        | `__add__()`     |
| -        | `__sub__()`     |
| *        | `__mul__()`     |
| /        | `__truediv__()` |

---

## Example

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```

### Output

```python
30
```

---

## Key Point

**Operator Overloading allows custom objects to behave like built-in data types.**

---

## Easy Remember

* `+` → `__add__()`
* `-` → `__sub__()`
* `*` → `__mul__()`
* `/` → `__truediv__()`

### One-Line Interview Answer

**Operator Overloading allows operators to work differently with user-defined objects by implementing special magic methods.**


# 6. Duck Typing 🦆

  - Duck Typing means:

    - "If it looks like a duck and behaves like a duck, then treat it as a duck."

    - In Python, we don't care about the object's class. We only care whether the object has the required method or behavior.

````python
class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")

def make_animal_speak(animal):
    animal.speak()


dog = Dog()
cat = Cat()

make_animal_speak(dog)
make_animal_speak(cat)
````

    Here, make_animal_speak() doesn't check whether the object is a Dog or Cat.

    It only checks if the object has a speak() method.

    This is Duck Typing.

# Simple Analogy of Duck Typing

Think of **animal** as an empty box.

```python
def make_animal_speak(animal):
    animal.speak()
```

### What Happens?

* Put a **Dog** in the box → it barks.
* Put a **Cat** in the box → it meows.

The function doesn't care what object is inside the box.

It only expects:

> **"Whatever you give me should have a `speak()` method."**

---

## Why is this Duck Typing?

Python checks whether the object can perform the required action, not what type it is.

```python
Dog → has speak() ✓
Cat → has speak() ✓
```

So both objects work with the same function.

---

## One-Line Interview Definition

**Duck Typing is a concept where Python focuses on an object's behavior (methods) rather than its actual type.**
