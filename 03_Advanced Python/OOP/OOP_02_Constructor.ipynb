## 1 self Keyword
  - What is self?

  - self refers to the current object.

  - When an object calls a method, Python automatically passes that object as the first argument.

````python
class student:

      def greet(self):
         print("Hello, I am a Student")

s1 = student()
s1.greet()
````

Internally Python does:

Student.greet(student1)

Here:

self = student1

 # Why self is Needed?

To access:

- instance variables
- instance methods

inside the class.

````python
class Student:

    def show(self):
        print("Welcome")

student1 = Student()

student1.show()
````

Without self, Python will give an error.

# 2.  Constructor (init)
    - What is a Constructor?
    - A constructor is a special method that runs automatically when an object is created.

````python
# Synatx of constructor

class Student:

    def __init__(self, name):              ## __init__() → Constructor
       print("Constructor called")
````

# Example 1: Simple Constructor

````python
class Student:

   def __init__(self):
       print("Object is created")

s1 = Student()   # Object is created
````

# Why Use Constructor?

To initialize:

 - name
 - age
 - salary
 - course
 - product details

immediately when an object is created.

# Example 2: Parameterized Constructor

````python
class Student:

    def __init__(self, name):
       self.name = name

s1 = Student("Om")
print(s1.name)

````

Understanding Step-by-Step

Object creation:

 - student1 = Student("Om")

Internally:

 - Student.__init__(student1, "Om")

Values become:

 - self = student1
 - name = Om

Then:

 - self.name = name

becomes:

 - student1.name = "Om"

# Example 3: Multiple Parameters 

````python
class Student:

    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

s1 = Student("Ram", 22, "Computer Science")
print(s1.name, s1.age, s1.course)
````

Multiple Objects

````python
class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age

student1 = Student("Om", 21)
student2 = Student("Rahul", 22)

print(student1.name)
print(student2.name)
````

# Constructor + Method

````python
class Student:

   def __init__(self, name):

      self.name = name

   def display(self):
      print("Name:", self.name)

s1 = Student("Om")
s2 = Student("Rahul")

s1.display()
s2.display()
````

# Constructor Flow

            Student("Om", 21)

                  ↓

            __init__(self, name, age)

                  ↓

            self.name = name
            self.age = age

                  ↓

            Object Initialized

# Constructor (**init**) - Interview Notes

## 1. What is a Constructor?

A constructor is a special method that is automatically called when an object is created.

Example:

```python
class Student:

    def __init__(self):
        print("Object Created")

student1 = Student()
```

---

## 2. What is the Name of Constructor in Python?

```python
__init__()
```

---

## 3. Why is a Constructor Used?

A constructor is used to initialize object data.

Example:

```python
class Student:

    def __init__(self, name):
        self.name = name
```

---

## 4. Is Constructor Called Automatically?

Yes.

```python
student1 = Student()
```

automatically calls:

```python
__init__()
```

---

## 5. Can a Constructor Accept Parameters?

Yes.

```python
class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age
```

---

## 6. What is a Parameterized Constructor?

A constructor that accepts parameters.

```python
class Student:

    def __init__(self, name):

        self.name = name
```

---

## 7. What is a Default Constructor?

A constructor without parameters (except self).

```python
class Student:

    def __init__(self):

        print("Student Created")
```

---

## 8. What is self in Constructor?

self refers to the current object.

```python
class Student:

    def __init__(self, name):

        self.name = name
```

---

## 9. Can a Constructor Return a Value?

No.

Wrong:

```python
class Student:

    def __init__(self):

        return 10
```

This gives TypeError.

---

## 10. Can a Class Exist Without a Constructor?

Yes.

```python
class Student:
    pass
```

Python provides a default constructor internally.

---

## 11. Difference Between Constructor and Method

| Constructor          | Method                |
| -------------------- | --------------------- |
| **init**()           | User-defined function |
| Called automatically | Called manually       |
| Initializes object   | Performs action       |

---

## 12. Can Python Have Multiple Constructors?

No.

Python does not support constructor overloading directly.

```python
class Student:

    def __init__(self):
        pass

    def __init__(self, name):
        pass
```

Only the last constructor will be used.
