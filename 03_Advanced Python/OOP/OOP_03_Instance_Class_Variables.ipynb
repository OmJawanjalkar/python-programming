# Instance Variables and Class Variables

## What are Variables in a Class?

Variables are used to store data inside a class.

Example:

* name
* age
* salary
* college

---

# 1. Instance Variables

## Definition

Instance variables belong to a specific object.

Each object gets its own copy of the variable.

---

## Example

```python
class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age
```

Here:

```python
self.name
self.age
```

are instance variables.

---

## Characteristics of Instance Variables

* Created using self
* Object-specific
* Multiple copies exist
* Values can be different for each object

---

## Example

```python
class Student:

    def __init__(self, name):

        self.name = name

student1 = Student("Om")
student2 = Student("Rahul")

print(student1.name)
print(student2.name)
```

Output:

```python
Om
Rahul
```

---

# 2. Class Variables

## Definition

Class variables belong to the class.

They are shared by all objects.

---

## Example

```python
class Student:

    college = "ABC College"
```

Here:

```python
college
```

is a class variable.

---

## Characteristics of Class Variables

* Defined directly inside class
* Shared by all objects
* Only one copy exists
* Same value for all objects

---

## Example

```python
class Student:

    college = "ABC College"

student1 = Student()
student2 = Student()

print(student1.college)
print(student2.college)
```

Output:

```python
ABC College
ABC College
```

---

# Instance Variable vs Class Variable

| Instance Variable             | Class Variable             |
| ----------------------------- | -------------------------- |
| Belongs to object             | Belongs to class           |
| Uses self                     | Defined directly in class  |
| Separate copy for each object | One shared copy            |
| Values can differ             | Same value for all objects |

---

# Real-Life Example

```python
class Student:

    college = "ABC College"

    def __init__(self, name):

        self.name = name
```

Instance Variable:

```python
name
```

Class Variable:

```python
college
```

---

# Complete Example

```python
class Student:

    college = "ABC College"

    def __init__(self, name, age):

        self.name = name
        self.age = age

student1 = Student("Om", 21)
student2 = Student("Rahul", 22)

print(student1.name)
print(student1.college)

print(student2.name)
print(student2.college)
```

Output:

```python
Om
ABC College

Rahul
ABC College
```

---

# Why Use Instance Variables?

* Store unique information
* Different values for different objects

Examples:

* name
* age
* salary

---

# Why Use Class Variables?

* Store common information
* Shared by all objects

Examples:

* college name
* company name
* bank name
