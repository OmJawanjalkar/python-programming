## 1. Basic Functions

A **function** is a reusable block of code that performs a specific task.  
You define a function using the `def` keyword.

```python
def function_name():
    # body of the function
    ...
```

````python
def my_function():
    print("Hello from the function!")

my_function()
````

````python
def Om_Info():
    print("My name is Om")
    print("I am 22 years old")
    print("I am a Data Scientist")

Om_Info()
````

---
## 2. Positional Arguments

**Positional arguments** are passed to a function in the order they are defined.  
The values are matched to the parameters by their **position**.

```python
def add(a, b):   # a gets 1st value, b gets 2nd
    ...
```

````python
def add(a, b):
    print(a + b)

add(5, 10)
````

````python
def Info(name, age):
    print("My name is : " + str(name))
    print("My age is :" + str(age))

Info("Aditi", 20)
Info("Payal", "22")
````

---
## 3. Keyword Arguments

**Keyword arguments** allow you to pass arguments by explicitly naming the parameter.  
This means the **order doesn't matter** — you can pass them in any sequence.

```python
Info(age=22, name="Aditi")  # order reversed, still works!
```

````python
def Info(name, age):
    print(name, age)

Info(age=22, name="Aditi")
````

---
## 4. Default Arguments

**Default arguments** allow a function to be called with fewer arguments than it is defined to allow.  
If no value is passed for that parameter, the **default value** is used.

```python
def City(name="Hyderabad"):  # "Hyderabad" is the default
    ...
```

````python
def City(name="Hyderabad"):
    print("I am from " + name)

City()             # uses default
City("Bangalore")  # overrides default
City("Amaravati")  # overrides default
````

---
## 5. Variable-Length Arguments: `*args` and `**kwargs`

Sometimes you don't know how many arguments will be passed. Python provides two special syntaxes:

| Syntax | Name | What it does |
|--------|------|--------------|
| `*args` | Non-keyword variable args | Collects extra **positional** arguments as a **tuple** |
| `**kwargs` | Keyword variable args | Collects extra **keyword** arguments as a **dict** |

**Use cases:**
- `*args` → Adding unlimited numbers: `1, 2, 3, 4, 5`
- `**kwargs` → Form details: `name="Om"`, `age=21`, `city="Nagpur"`

````python
# *args — variable number of positional arguments
def my_function(*args):
    return sum(args)

result = my_function(10, 20, 30, 40)
print(result)
````

````python
# Combining *args and **kwargs
def demo(*args, **kwargs):
    print("Positional Arguments: ", args)
    print("Keyword Arguments:   ", kwargs)

demo(10, 20, 30, name="Ramkrushna", age=56)
````

## 6. Recursion

**Recursion** is when a function **calls itself** to solve a smaller version of the same problem.

Every recursive function needs:
1.  A **base case** — the condition that stops the recursion
2.  A **recursive case** — where the function calls itself

> ⚠️ Without a base case, recursion will run forever and cause a `RecursionError`.

**How it works with `show(5)`:**
```
show(5) → print 5 → show(4)
show(4) → print 4 → show(3)
show(3) → print 3 → show(2)
show(2) → print 2 → show(1)
show(1) → print 1 → show(0)
show(0) → return  (base case reached)
```

````python
def show(n):
    if n == 0:      # base case
        return
    print(n)        # recursive case
    show(n - 1)     # function calls itself

show(5)
````
