# Introduction to Python

## What is Python?

Python is a high-level, interpreted, and general-purpose programming language. It was created by Guido van Rossum and released in 1991. Python is known for its simple syntax, readability, and versatility, making it one of the most popular programming languages in the world.

---

## Features of Python

### 1. Easy to Learn and Read
Python uses simple English-like syntax, making it beginner-friendly.

### 2. Interpreted Language
Python code is executed line by line, which makes debugging easier.

### 3. Object-Oriented
Python supports object-oriented programming concepts such as classes, objects, inheritance, and polymorphism.

### 4. Cross-Platform
Python programs can run on Windows, Linux, and macOS without modification.

### 5. Open Source
Python is free to use and has a large community of developers.

### 6. Extensive Libraries
Python provides thousands of libraries for web development, data science, machine learning, automation, and more.

## Python Execution Process

1. Write Python code.
2. Python Interpreter converts the code into bytecode.
3. Python Virtual Machine (PVM) executes the bytecode.
4. Output is generated.
