# Python Loops

## What are Loops?

Loops are used to execute a block of code repeatedly until a specified condition is met.

Python provides two types of loops:

1. While Loop
2. For Loop

---

# While Loop

A while loop repeatedly executes a block of code as long as the given condition is True.

### Syntax

```python
while condition:
    # code block
```

### Example: Print Numbers from 1 to 5

````python
count = 1
while count <= 5:
  print(count)
  count += 1
````

### Example: Print Numbers from 5 to 1

````python
count = 5

while count >= 1:
  print(count)
  count -= 1
````

# Applications of While Loop

### Print Numbers from 1 to 100

````python
count = 1
while count <= 100:
  print(count)
  count += 1
````

### Print Numbers from 100 to 1

````python
count = 100

while count >= 1:
  print(count)
  count -= 1
  
print("OM")
````

## For Loop 

# 🔁 What is a `for` Loop?

A **`for` loop** is used to iterate over a sequence and execute a block of code **once for each item** in that sequence.

> **Definition:**  
> A `for` loop executes a block of code once for each item in an iterable.

---

# 📝 Syntax

```python
for variable in iterable:
    # Code to execute
```

---

# 🔍 Syntax Breakdown

- **`variable`** → Holds the current item during each iteration.
- **`iterable`** → A collection of items that can be iterated over.

---

## Example 1: Loop through a List

````python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
  print(fruit)
````

### Example 2: Loop through a String

````python
name = "Python"

for letter in name:
    print(letter)
````

### Example 3: Using range()

    The range() function generates a sequence of numbers.

````python
for i in range(5):
    print(i)
````

### 2. range(start, stop)

````python
for i in range(2, 6):
    print(i)
````

### 3. range(start, stop, step)

````python
for i in range(1, 10, 2):
    print(i)
````

#### Reverse Loop

````python
for i in range(10, 0, -1):
    print(i)
````

### Example 4: Sum of Numbers

````python
total = 0

for i in range(1, 6):
    total = total + i

print(total)
````

### Example 5: Multiplication Table

````python
number = 5

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
````

### Example 6: Nested for Loop

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
````

# 🎯 Common Interview Questions – `for` Loop in Python

These are some of the **most frequently asked interview questions** on Python `for` loops in technical interviews.

---

## 📌 1. What is a `for` loop in Python?

A `for` loop is used to iterate over an iterable (such as a list, tuple, string, dictionary, or range) and execute a block of code once for each item.

---

## 📌 2. What is an iterable?

An **iterable** is any object that can be traversed one element at a time using a loop.

**Examples:**
- List
- Tuple
- String
- Dictionary
- Set
- `range()`

---

## 📌 3. How does `range()` work?

The `range()` function generates a sequence of numbers and is commonly used with `for` loops.

**Syntax:**
```python
range(start, stop, step)
```

**Example:**
```python
for i in range(1, 6):
    print(i)
```

**Output:**
```
1
2
3
4
5
```

---

## 📌 4. What is the difference between `break` and `continue`?

| `break` | `continue` |
|----------|------------|
| Terminates the loop immediately. | Skips the current iteration and moves to the next one. |

---

## 📌 5. What is the purpose of `pass`?

The `pass` statement is a placeholder that does nothing. It is used when a statement is syntactically required but no action is needed.

**Example:**
```python
for i in range(5):
    pass
```

---

## 📌 6. Can a `for` loop iterate over a dictionary?

**Yes.** A `for` loop can iterate over:
- Keys
- Values
- Key-value pairs

---

## 📌 7. What is a nested `for` loop?

A nested `for` loop is a `for` loop inside another `for` loop.

**Example:**
```python
for i in range(3):
    for j in range(2):
        print(i, j)
```

---

## 📌 8. What is the difference between `for` and `while` loops?

| `for` Loop | `while` Loop |
|------------|--------------|
| Used when the number of iterations is known. | Used when the number of iterations is unknown. |
| Iterates over an iterable. | Continues until a condition becomes `False`. |

---

## 📌 9. When does the `else` block execute in a `for` loop?

The `else` block executes **only if the loop completes normally**, without encountering a `break` statement.

---

## 📌 10. What is the time complexity of a simple `for` loop?

If the loop iterates **n** times and performs a constant-time operation in each iteration, the time complexity is:

```text
O(n)
```

Where:
- **n** = Number of elements in the iterable.

---
