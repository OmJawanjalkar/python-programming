# Python Data Types

## Introduction

Data types specify the type of value that a variable can store. Python provides several built-in data types that help organize and manipulate data efficiently.

---

## Numeric Data Types

### Integer (int)

Integers are whole numbers without a decimal point.

````python
x = 10
y = -25
````

### Float (float)

Floats are numbers with decimal points.

````python
pi = 3.14
price = 99.99
````

### Complex (complex)

Complex numbers contain a real and an imaginary part.

````python
z = 2 + 3j
````

## Text Data Type

### String (str)

Strings are used to store textual data and are enclosed within single or double quotes.


````python
srt = "Om"

name = "Boob"

message = 'Hello World'
````



---

## Sequence Data Types

### List (list)

A list is an ordered and mutable collection of items.

```python
fruits = ["Apple", "Banana", "Mango"]
```

**Characteristics:**
- Ordered
- Mutable (can be modified)
- Allows duplicate values

### Tuple (tuple)

A tuple is an ordered and immutable collection of items.

```python
coordinates = (10, 20)
```

**Characteristics:**
- Ordered
- Immutable (cannot be modified)
- Allows duplicate values

### Range (range)

Used to generate a sequence of numbers.

```python
numbers = range(1, 6)
```

---

## Set Data Types

### Set (set)

A set is an unordered collection of unique elements.

```python
colors = {"Red", "Green", "Blue"}
```

**Characteristics:**
- Unordered
- No duplicate values
- Mutable

### Frozen Set (frozenset)

An immutable version of a set.

```python
fset = frozenset([1, 2, 3])
```

---

## Mapping Data Type

### Dictionary (dict)

A dictionary stores data in key-value pairs.

```python
student = {
    "name": "John",
    "age": 20
}
```

**Characteristics:**
- Stores key-value pairs
- Mutable
- Keys must be unique

---

## Boolean Data Type

### Boolean (bool)

Boolean values represent True or False.

```python
is_student = True
is_logged_in = False
```

---

## None Data Type

### NoneType

Represents the absence of a value.

```python
data = None
```

---

## Checking Data Types

The `type()` function is used to determine the data type of a variable.

```python
x = 10
print(type(x))
```

**Output:**

```python
<class 'int'>
```
