# Sets in Python

A **set** is a built-in Python data structure used to store an **unordered collection of unique elements**. Sets are enclosed within curly braces `{}` and automatically remove duplicate values.

Sets are **mutable**, allowing elements to be added or removed after creation. However, individual elements must be immutable. Sets are optimized for fast membership testing and support mathematical operations such as union, intersection, difference, and symmetric difference.

Sets are widely used in **Data Science, Machine Learning, database operations, duplicate removal, and data analysis tasks**.

## Key Characteristics

- Unordered collection.
- Stores only unique elements.
- Mutable data structure.
- Does not support indexing or slicing.
- Automatically removes duplicates.
- Fast membership testing.
- Supports set operations such as Union, Intersection, Difference, and Symmetric Difference.

> **Example:** `{10, 20, 30, 40}` is a set containing unique elements.

### Set Creation

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("A =", A)
print("B =", B)
```

### Add Element

```python
A.add(10)

print("After Add")
print(A)
```

### Update Set

```python
A.update([20, 30])

print("After Update")
print(A)
```

### Remove Element

```python
A.remove(2)

print("After Remove")
print(A)
```

### Discard Element

```python
A.discard(100)

print("After Discard")
print(A)
```

### Pop Element

```python
A.pop()

print("After Pop")
print(A)
```

### Length

```python
print("Length")
print(len(A))
```

### Membership Operator

```python
print("Check Element")
print(4 in A)
```

### Union

```python
print("Union")
print(A | B)
```

### Intersection

```python
print("Intersection")
print(A & B)
```

### Difference

```python
print("Difference")
print(A - B)
```

### Symmetric Difference

```python
print("Symmetric Difference")
print(A ^ B)
```

### Loop Through Set

```python
print("Loop")

for x in A:
    print(x)
```

### Copy Set

```python
C = A.copy()

print("Copied Set")
print(C)
```

### Subset

```python
X = {1, 2}
Y = {1, 2, 3, 4}

print("Subset")
print(X.issubset(Y))
```

### Superset

```python
print("Superset")
print(Y.issuperset(X))
```

### Remove Duplicates

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique = set(numbers)

print("Unique Values")
print(unique)
```

### Convert Set to List

```python
data = list(unique)

print("Converted List")
print(data)
```

### Frozen Set

```python
f = frozenset([1, 2, 3])

print("Frozen Set")
print(f)
```

### Clear Set

```python
B.clear()

print("After Clear")
print(B)
```
