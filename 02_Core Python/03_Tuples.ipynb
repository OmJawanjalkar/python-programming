# Tuples in Python

A **tuple** is a built-in Python data structure used to store **an ordered collection of items**. Tuples are enclosed within parentheses `()` and can contain elements of different data types.

Unlike lists, tuples are **immutable**, meaning their elements cannot be modified after creation. Because of their immutability, tuples are faster than lists and are commonly used for storing fixed or read-only data.

Tuples are widely used in **function returns, database records, coordinate systems, configuration settings, and data processing applications**.

## Key Characteristics

- Ordered collection.
- Immutable (cannot be modified).
- Allows duplicate elements.
- Supports indexing and slicing.
- Can store heterogeneous data types.
- Faster than lists for read-only operations.
- Can be used as dictionary keys if all elements are immutable.

> **Example:** `(101, "Om", 8.5)` is a tuple containing different types of data.

### Tuple Creation

```python
Name = ("Om", "Pratik", "Satyarth", "Anshul")
print(Name)
```

### Accessing Tuple Elements

```python
print(Name[0])
print(Name[1])   # Pratik
print(Name[-1])  # Anshul
```

### Tuple Slicing

```python
print(Name[1:3])
```

### Reversing a Tuple

Tuples are immutable, so we cannot modify them directly — but we can create a reversed copy using slicing.

```python
print(Name[::-1])  # Reverse the tuple
```

### Mixed Data Types

Tuples can contain different data types together.

```python
MixedTuple = ("Hello", 42, 3.14, True)
print(MixedTuple)
```

### Tuple Concatenation

```python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

print(tuple1 + tuple2)  # (1, 2, 3, 4, 5, 6)
```

### Length of a Tuple

```python
print(len(Name))
```

### Repetition of Tuples

```python
print(tuple1 * 3)  # (1, 2, 3, 1, 2, 3, 1, 2, 3)
```

### Membership Operator

```python
print(2 in tuple1)    # True
print("om" in Name)   # False (case-sensitive)
print("Om" in Name)   # True
```

### Counting Occurrences (count())

```python
data = (1, 2, 2, 2)

print(data.count(2))  # 3
print(data.count(5))  # 0
```

### Finding Index (index())

```python
print(Name.index("Pratik"))
print(tuple1.index(2))
```

### Nested Tuples

```python
data = ((1, 2), (3, 4))

print(data[1][0])  # 3
```

### Tuple Packing and Unpacking

**Packing** stores multiple values into a single tuple. **Unpacking** extracts those values into separate variables.

```python
# Packing
person = ("Alice", 30, "Engineer")

# Unpacking
name, age, profession = person

print(name)         # Alice
print(age)          # 30
print(profession)   # Engineer
```

### Converting a List to a Tuple

```python
my_list = [1, 2, 3, 4]
my_tuple = tuple(my_list)

print(my_tuple)  # (1, 2, 3, 4)
```

### Converting a Tuple to a List

```python
my_tuple = (1, 2, 3, 4)
my_list = list(my_tuple)

print(my_list)  # [1, 2, 3, 4]
```

### Modifying a Tuple (via List Conversion)

Since tuples are immutable, to change a value we convert to a list, modify it, then convert back to a tuple.

```python
numbers = (10, 20, 30)

temp = list(numbers)
temp[0] = 100

numbers = tuple(temp)
print(numbers)  # (100, 20, 30)
```

### Maximum, Minimum, and Sum

```python
values = (5, 2, 9, 1)

print(max(values))  # 9
print(min(values))  # 1
print(sum(values))  # 17
```

### Looping Through a Tuple

```python
print("Looping through Tuple:")
for v in values:
    print(v)
```

### Deleting a Tuple

Individual elements cannot be deleted (tuples are immutable), but the entire tuple object can be deleted using `del`.

```python
college = ("MIT", "Stanford", "Harvard")

del college

try:
    print(college)
except NameError as e:
    print("Error:", e)
```
