# Python Lists

A quick, hands-on notebook covering the essentials of Python lists.

## Theory

- A **list** is an ordered, mutable collection of items.
- Lists are defined using square brackets `[]`.
- Lists can hold items of different data types, including other lists (nested lists).
- Items are accessed using **indexing** (from 0) and **slicing**.
- Lists support duplicates, unlike sets.
- Python provides many built-in methods to add, remove, sort, and search list items.

### List Creation

```python
A = [1, 2, 3, 4]
B = [5, 6, 7, 8]

print("A =", A)
print("B =", B)
```

### Indexing

```python
print("First Element")
print(A[0])

print("Last Element")
print(A[-1])
```

### Slicing

```python
print("Slice A[1:3]")
print(A[1:3])

print("Reversed A")
print(A[::-1])
```

### Length

```python
print("Length")
print(len(A))
```

### Append Element

```python
A.append(10)

print("After Append")
print(A)
```

### Insert Element

```python
A.insert(1, 100)

print("After Insert")
print(A)
```

### Extend List

```python
A.extend([20, 30])

print("After Extend")
print(A)
```

### Remove Element by Value

```python
A.remove(100)

print("After Remove")
print(A)
```

### Pop Element

```python
popped = A.pop()

print("Popped Value")
print(popped)

print("After Pop")
print(A)
```

### Delete Element by Index

```python
del A[0]

print("After Delete")
print(A)
```

### Update Element

```python
A[0] = 999

print("After Update")
print(A)
```

### Membership Operator

```python
print("Check Element")
print(4 in A)
```

### Concatenation

```python
C = A + B

print("Concatenated List")
print(C)
```

### Repetition

```python
print("Repeated List")
print(B * 2)
```

### Count Occurrences

```python
nums = [1, 2, 2, 3, 4, 4, 4, 5]

print("Count of 4")
print(nums.count(4))
```

### Find Index of Element

```python
print("Index of 3")
print(nums.index(3))
```

### Sort List

```python
nums.sort()

print("Sorted List")
print(nums)
```

### Sort in Descending Order

```python
nums.sort(reverse=True)

print("Descending List")
print(nums)
```

### Sorted Copy (Original Unchanged)

```python
original = [5, 3, 1, 4, 2]
new_sorted = sorted(original)

print("Original")
print(original)

print("Sorted Copy")
print(new_sorted)
```

### Reverse List

```python
original.reverse()

print("Reversed List")
print(original)
```

### Copy List

```python
D = A.copy()

print("Copied List")
print(D)
```

### Clear List

```python
E = [1, 2, 3]
E.clear()

print("Cleared List")
print(E)
```

### Loop Through List

```python
print("Loop")
for item in B:
    print(item)
```

### Loop with Index (enumerate)

```python
print("Loop with Index")
for i, item in enumerate(B):
    print(i, item)
```

### Nested Lists

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print("Matrix")
print(matrix)

print("Element at row 1, col 2")
print(matrix[1][2])
```

### List Comprehension

```python
squares = [x**2 for x in range(1, 6)]

print("Squares")
print(squares)
```

```python
evens = [x for x in range(1, 11) if x % 2 == 0]

print("Even Numbers")
print(evens)
```

### Min, Max, Sum

```python
values = [10, 20, 5, 40, 15]

print("Minimum")
print(min(values))

print("Maximum")
print(max(values))

print("Sum")
print(sum(values))
```

### Remove Duplicates

```python
duplicates = [1, 2, 2, 3, 4, 4, 5]
unique = list(set(duplicates))

print("Unique Values")
print(unique)
```

### Convert List to String

```python
words = ["Python", "is", "fun"]
sentence = " ".join(words)

print("Sentence")
print(sentence)
```
