# Lambda Functions in Python

## What is a Lambda Function?

A **Lambda Function** is a small, anonymous (unnamed) function in Python that is created using the **`lambda`** keyword. Unlike a normal function defined using the `def` keyword, a lambda function does not require a function name and automatically returns the result of its expression.

A lambda function can accept **zero, one, or multiple arguments**, but it can contain **only one expression**. Because of its concise syntax, it is commonly used for short, simple operations where defining a complete function would be unnecessary.


---

## Characteristics of Lambda Functions

- Anonymous function (does not require a function name).
- Created using the `lambda` keyword.
- Can take any number of arguments.
- Contains only one expression.
- Automatically returns the result of the expression.
- Mainly used for small and temporary functions.

---

## Syntax

```python
lambda arguments: expression
```

### Syntax Explanation

- **lambda** → Keyword used to create the anonymous function.
- **arguments** → Input parameters passed to the function.
- **:** → Separates the arguments from the expression.
- **expression** → A single expression that is evaluated and returned automatically.

---




### Example 1: Lambda Function with One Argument

````python
square_lambda = lambda x: x**x
print(square_lambda(2))
````

### Example 2: Lambda Function with Multiple Arguments

````python
Addition = lambda a,b : a + b
print(Addition(2,3))
````

###  Example 3: Lambda Function Without Arguments

````python
greet = lambda : "welcome to OM repo"
print(greet())
````

### Example 4: Lambda Function with Default Argument

````python
multiply = lambda x, y=6: x*y
print(multiply(3))
````

### Example 5: Lambda Function Using if-else

````python
maximum = lambda a,b : a if a>b else b
print(maximum(33,55))
````

### Example 6: Check Even or Odd

````python
check = lambda x: "Even" if x %2 == 0 else "Odd"
print(check(55))
````

## Flow of Execution

1. Python reads the `lambda` keyword.
2. The input arguments are received.
3. The expression is evaluated.
4. The computed value is automatically returned.
5. No `return` statement is required.

---

## Difference Between Normal Function and Lambda Function

| Normal Function | Lambda Function |
|-----------------|-----------------|
| Defined using `def` | Defined using `lambda` |
| Requires a function name | Anonymous (can be assigned to a variable) |
| Can contain multiple statements | Can contain only one expression |
| Uses the `return` keyword | Returns automatically |
| Suitable for complex logic | Suitable for simple operations |

---

## When Should We Use Lambda Functions?

Lambda functions are best used when:

- The function is very small.
- The function is used only once.
- A built-in function requires another function as an argument.
- The logic consists of a single expression.

Avoid using lambda functions for complex business logic because normal functions are easier to read and maintain.
