# Higher-Order  Functions in Python: `map()`, `filter()`, `reduce()`, and `sorted()`

## Introduction

Higher-order functions are functions that either accept another function as an argument or return a function. In Python, functions such as **`map()`**, **`filter()`**, **`reduce()`**, and **`sorted()`** are called higher-order functions because they can work with other functions, often lambda functions.

These functions simplify data processing, reduce the amount of code, and improve readability. They are widely used in data analysis, machine learning, and functional programming.

---

# 1. `map()` Function

## What is `map()`?

The `map()` function applies a specified function to every element of an iterable (such as a list, tuple, or set) and returns a **map object (iterator)**. To display the result, the map object is usually converted into a list.

### Syntax

```python
map(function, iterable)
```

### Parameters

* **function** – Function to apply to every element.
* **iterable** – List, tuple, set, or any iterable object.

### Return Value

Returns a **map object** containing the transformed values.


### Example 1

````python
names = ["om", "rahul", "amit"]

result = map(str.upper, names)

print(list(result))
````

### Example 2

````python
numbers = ["10", "20", "30", "40"]

result = map(int, numbers)

print(list(result))
````

### Example 3


````python
numbers = [1, 2, 3, 4, 5]

square = list(map(lambda x: x**2, numbers))
print(square)
````

### Explanation

* `lambda x: x ** 2` calculates the square of each number.
* `map()` applies this function to every element in the list.
* `list()` converts the returned map object into a list.

### Applications

* Data transformation
* Unit conversion
* Feature engineering
* String formatting

# 2. `filter()` Function

## What is `filter()`?

The `filter()` function selects only those elements from an iterable for which a given function returns **True**.

### Syntax

```python
filter(function, iterable)
```

### Parameters

* **function** – Function that returns `True` or `False`.
* **iterable** – Collection of elements.

### Return Value

Returns a **filter object** containing only the matching elements.

### Example


````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
````

````python
numbers = [-5, 10, -15, 20, -25, 30]

result = filter(lambda x: x > 0, numbers)

print(list(result))
````

````python
words = ["Python", "AI", "Machine", "Data", "Learning"]

result = filter(lambda word: len(word) > 5, words)

print(list(result))
````

### Explanation

* The lambda function checks whether a number is even.
* If the condition is `True`, the element is kept.
* If the condition is `False`, the element is discarded.

### Applications

* Removing unwanted data
* Data cleaning
* Selecting records
* Filtering datasets

# 3. `reduce()` Function

## What is `reduce()`?

The `reduce()` function repeatedly applies a function to the elements of an iterable and combines them into a **single value**. It is available in Python's **`functools`** module.

### Syntax

```python
from functools import reduce

reduce(function, iterable)
```

### Parameters

* **function** – Function that combines two values.
* **iterable** – Collection of elements.

### Return Value

Returns a **single accumulated value**.

### Example


````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
total = reduce(lambda x, y: x + y, numbers)
print(total)
````

### Step-by-Step Execution

```
1 + 2 = 3
3 + 3 = 6
6 + 4 = 10
10 + 5 = 15
```

### Explanation

* `reduce()` starts with the first two elements.
* It combines them using the lambda function.
* The result is combined with the next element.
* The process continues until one final value remains.

### Applications

* Sum of numbers
* Product of numbers
* Finding maximum or minimum
* Aggregation operations

# 4. `sorted()` Function

## What is `sorted()`?

The `sorted()` function returns a **new sorted list** from any iterable. It does **not modify** the original data.

### Syntax

```python
sorted(iterable, key=None, reverse=False)
```

### Parameters

* **iterable** – Data to sort.
* **key** – Function used to determine the sorting order.
* **reverse** – If `True`, sorts in descending order.

### Return Value

Returns a **new sorted list**.

### Example

````python
students = [
    ("John", 85),
    ("Alice", 95),
    ("Bob", 78)
]

sorted_students = sorted(
    students,
    key=lambda student: student[1]
)

print(sorted_students)
````

### Explanation

* Each tuple contains `(Name, Marks)`.
* The lambda function returns the marks (`student[1]`).
* `sorted()` uses these values to arrange the data in ascending order.

### Descending Order

````python
sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print(sorted_students)
````

### Applications

* Sorting by marks
* Sorting dictionaries
* Ranking predictions
* Ordering datasets

# Comparison of Higher-Order Functions

| Function   | Purpose                                   | Return Type   |
| ---------- | ----------------------------------------- | ------------- |
| `map()`    | Applies a function to every element       | Map object    |
| `filter()` | Keeps elements that satisfy a condition   | Filter object |
| `reduce()` | Combines all elements into a single value | Single value  |
| `sorted()` | Returns a new sorted list                 | List          |

---

# Summary

* **`map()`** transforms every element of an iterable.
* **`filter()`** selects elements that satisfy a condition.
* **`reduce()`** combines all elements into one final value.
* **`sorted()`** returns a new sorted version of the data.

These functions are essential in Python because they make code shorter, more readable, and more efficient. They are widely used in data science, machine learning, automation, and functional programming.
