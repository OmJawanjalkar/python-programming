# Python Dictionaries

A quick, hands-on notebook covering the essentials of Python dictionaries.

## 1. Create a Dictionary

```python
my_dict = {"name": "John", "age": 30, "city": "New York"}
print(my_dict)
```

## 2. Accessing Values

```python
print(my_dict["name"])
print(my_dict["city"])
```

## 3. Adding and Updating Data

```python
# Update existing key
my_dict["age"] = 31
print(my_dict)
```

```python
# Add new key-value pair
my_dict["mail"] = "Omjawanjalkar@gmail.com"
print(my_dict)
```

## 4. Removing Key-Value Pairs

```python
del my_dict["city"]
print(my_dict)
```

## 5. Dictionary Methods

Common built-in methods:
- `keys()`
- `values()`
- `items()`
- `get()`
- `update()`
- `pop()`
- `clear()`

## 6. Looping Through a Dictionary

```python
print("Looping through Dictionary:")
for key in my_dict:
    print(f"{key}: {my_dict[key]}")
```

## 7. Nested Dictionaries

```python
nested_dict = {
    "person1": {"name": "Alice", "age": 30},
    "person2": {"name": "Bob", "age": 25}
}
print(nested_dict)
```

## 8. Dictionary Comprehension

```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```

```python
check_num = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 11)}
print(check_num)
```

## 9. Checking if a Key Exists

```python
if "name" in my_dict:
    print("Name exists in the dictionary.")
else:
    print("Name does not exist in the dictionary.")
```
