# Dictionary in Python

A **dictionary** is one of Python's most powerful built-in data structures used to store data in **key-value pairs**. Each key in a dictionary is unique and is associated with a specific value, making it easy to organize, access, and update data efficiently.

Unlike lists and tuples, which use numerical indexing, dictionaries use **keys** to retrieve values. Python dictionaries are implemented using **hash tables**, allowing average **O(1)** time complexity for lookup, insertion, and deletion operations.

Dictionaries are **mutable**, meaning their contents can be modified after creation by adding, updating, or removing key-value pairs. They can store values of different data types, including numbers, strings, lists, tuples, and even other dictionaries.

Because of their fast lookup performance and flexible structure, dictionaries are extensively used in **Data Science, Machine Learning, Web Development, APIs, JSON data processing, configuration management, and database-like applications**.

## Key Characteristics

- Stores data as **key-value pairs**.
- Each **key must be unique**.
- Keys must be **immutable** (e.g., strings, integers, tuples).
- Values can be of **any data type**.
- Dictionaries are **mutable**, allowing modification after creation.
- Preserves insertion order (Python 3.7 and later).
- Provides **average O(1)** time complexity for search, insertion, and deletion operations.

> **Example:** A student's information can be stored as:
>
> - **Name** → Om
> - **Age** → 22
> - **Course** → Data Science
>
> Here, **Name**, **Age**, and **Course** are the **keys**, while **Om**, **22**, and **Data Science** are their corresponding **values**.

### 1. Create a Dictionary

```python
my_dict = {"name": "John", "age": 30, "city": "New York"}
print(my_dict)
```

### 2. Accessing Values

```python
print(my_dict["name"])
print(my_dict["city"])
```

### 3. Adding and Updating Data

```python
# Update existing key
my_dict["age"] = 31
print(my_dict)
```

```python
# Add new key-value pair
my_dict["mail"] = "Omjawanjalkar@gmail.com"
print(my_dict)
```

### 4. Removing Key-Value Pairs

```python
del my_dict["city"]
print(my_dict)
```

### 5. Dictionary Methods

Common built-in methods:
- `keys()`
- `values()`
- `items()`
- `get()`
- `update()`
- `pop()`
- `clear()`

### 6. Looping Through a Dictionary

```python
print("Looping through Dictionary:")
for key in my_dict:
    print(f"{key}: {my_dict[key]}")
```

### 7. Nested Dictionaries

```python
nested_dict = {
    "person1": {"name": "Alice", "age": 30},
    "person2": {"name": "Bob", "age": 25}
}
print(nested_dict)
```

### 8. Dictionary Comprehension

```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```

```python
check_num = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 11)}
print(check_num)
```

### 9. Checking if a Key Exists

```python
if "name" in my_dict:
    print("Name exists in the dictionary.")
else:
    print("Name does not exist in the dictionary.")
```
