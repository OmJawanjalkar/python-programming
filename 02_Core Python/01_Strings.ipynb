# Python Strings

A quick, hands-on notebook covering the essentials of Python strings.

## Theory

- A **string** is a sequence of characters enclosed in single `' '`, double `" "`, or triple `''' '''` quotes.
- Strings are **ordered** — each character has a fixed index position.
- Strings are **immutable** — once created, their content cannot be changed in place.
- Strings support **indexing** and **slicing** to access parts of the text.
- Python provides many built-in string methods for searching, formatting, and transforming text.
- Strings can be combined using **concatenation** (`+`) and **repetition** (`*`).
- **f-strings** allow embedding expressions directly inside string literals.

### String Creation

```python
s1 = "Hello, World!"
s2 = 'Python Programming'

print("s1 =", s1)
print("s2 =", s2)
```

### Indexing

```python
print("Indexing")
print(s1[0])
print(s1[-1])
```

### Slicing

```python
print("Slicing")
print(s1[0:5])
print(s1[7:])
print(s1[::-1])
```

### Length

```python
print("Length")
print(len(s1))
```

### Concatenation

```python
print("Concatenation")
print(s1 + " " + s2)
```

### Repetition

```python
print("Repetition")
print(s2 * 2)
```

### Membership Operator

```python
print("Check Substring")
print("Python" in s2)
```

### Change Case

```python
print("Upper Case")
print(s1.upper())

print("Lower Case")
print(s1.lower())
```

### Strip Whitespace

```python
s3 = "   Extra Spaces   "

print("Before Strip")
print(repr(s3))

print("\nAfter Strip")
print(repr(s3.strip()))
```

### Replace

```python
print("Replace")
print(s1.replace("World", "Python"))
```

### Split

```python
print("Split")
words = s2.split()
print(words)
```

### Join

```python
print("Join")
joined = "-".join(words)
print(joined)
```

### Find and Index

```python
print("Find")
print(s1.find("World"))

print("\nIndex")
print(s1.index("World"))
```

### Count

```python
print("Count")
print(s1.count("l"))
```

### startswith / endswith

```python
print("Starts With")
print(s1.startswith("Hello"))

print("Ends With")
print(s1.endswith("!"))
```

### String Formatting (f-strings)

```python
name = "Om"
age = 22

print("Formatted String")
print(f"My name is {name} and I am {age} years old.")
```

### Loop Through String

```python
print("Loop")
for ch in "Python":
    print(ch)
```

### Check Alphanumeric / Digit / Alpha

```python
print("isalpha")
print("Python".isalpha())

print("isdigit")
print("12345".isdigit())

print("isalnum")
print("Python3".isalnum())
```

### Immutability of Strings

```python
print("Immutability")
s4 = "Hello"
s5 = s4.replace("H", "J")

print("Original:", s4)
print("New:", s5)
```
